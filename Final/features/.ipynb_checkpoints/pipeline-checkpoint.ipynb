# Structure

```python
FEATURE_NOTEBOOK_VERSION = "fe_notebook_v0_1"

DEFAULT_SOURCE_ORDER = ("naip", "als", "3dep", "rap")

DEFAULT_FAMILY_ORDER = {
    "naip": ("raw", "veg_idx", "texture", "multiscale"),
    "als": ("height", "structure", "canopy_context"),
    "3dep": ("terrain", "terrain_context"),
    "rap": ("prior",),
}


@dataclass
class NotebookFEConfig:
    sites: list[str] = field(default_factory=lambda: list(cfg.sites))
    canonical_grid_source: str = cfg.features.canonical_grid_source
    object_table_enabled: bool = cfg.features.object_table_enabled

    source_order: tuple[str, ...] = DEFAULT_SOURCE_ORDER
    enabled_sources: dict[str, bool] = field(
        default_factory=lambda: {name: True for name in DEFAULT_SOURCE_ORDER}
    )
    enabled_families: dict[str, tuple[str, ...]] = field(
        default_factory=lambda: {k: tuple(v) for k, v in DEFAULT_FAMILY_ORDER.items()}
    )

    resampling_by_source: dict[str, Resampling] = field(
        default_factory=lambda: {
            "naip": Resampling.nearest,
            "als": Resampling.bilinear,
            "3dep": Resampling.bilinear,
            "rap": Resampling.bilinear,
        }
    )

    cache_root: Path = cfg.output.features_root / "notebook_cache"
    summary_root: Path = cfg.output.features_root / "notebook_summaries"
    qa_root: Path = cfg.output.features_root / "notebook_qa"

    force_refresh_assets: bool = False
    force_refresh_features: bool = False
    force_refresh_objects: bool = False

    version: str = FEATURE_NOTEBOOK_VERSION

    def resolve(self) -> "NotebookFEConfig":
        self.cache_root = Path(self.cache_root).resolve()
        self.summary_root = Path(self.summary_root).resolve()
        self.qa_root = Path(self.qa_root).resolve()

        for root in [self.cache_root, self.summary_root, self.qa_root]:
            root.mkdir(parents=True, exist_ok=True)

        return self


@dataclass(frozen=True)
class SourceSpec:
    name: str
    native_kind: str
    family_names: tuple[str, ...]
    default_resampling: Resampling
    notes: str = ""


@dataclass
class CanonicalGrid:
    width: int
    height: int
    transform: Affine
    crs: Any
    source_name: str
    nodata: float | int | None = None

    @property
    def pixel_size(self) -> tuple[float, float]:
        return abs(self.transform.a), abs(self.transform.e)

    def base_profile(self, *, dtype: str = "float32", count: int = 1, nodata: float | int | None = None) -> dict[str, Any]:
        return {
            "driver": "GTiff",
            "width": self.width,
            "height": self.height,
            "count": count,
            "dtype": dtype,
            "crs": self.crs,
            "transform": self.transform,
            "nodata": self.nodata if nodata is None else nodata,
        }


@dataclass
class SourceRasterBundle:
    site_id: str
    source_name: str
    arrays: dict[str, np.ndarray]
    transform: Affine
    crs: Any
    nodata: float | int | None = None
    metadata: dict[str, Any] = field(default_factory=dict)

    @property
    def shape(self) -> tuple[int, int]:
        first = next(iter(self.arrays.values()))
        return first.shape


@dataclass
class FeatureFamilyResult:
    site_id: str
    source_name: str
    family_name: str
    arrays: dict[str, np.ndarray]
    metadata: dict[str, Any] = field(default_factory=dict)


@dataclass
class SiteFeatureRun:
    site_id: str
    canonical_grid: CanonicalGrid | None = None
    source_bundles: dict[str, SourceRasterBundle] = field(default_factory=dict)
    family_results: dict[str, FeatureFamilyResult] = field(default_factory=dict)
    aligned_layers: dict[str, np.ndarray] = field(default_factory=dict)
    notes: list[str] = field(default_factory=list)


fe_cfg = NotebookFEConfig().resolve()
print(fe_cfg)
```

```python
def stable_json(payload: dict[str, Any]) -> str:
    return json.dumps(payload, sort_keys=True, default=str)


def hash_payload(payload: dict[str, Any]) -> str:
    return hashlib.sha1(stable_json(payload).encode("utf-8")).hexdigest()[:12]


def ensure_dir(path: str | Path) -> Path:
    path = Path(path)
    path.mkdir(parents=True, exist_ok=True)
    return path


def notebook_config_signature(notebook_cfg: NotebookFEConfig) -> str:
    return hash_payload(asdict(notebook_cfg))


def stage_cache_dir(
    notebook_cfg: NotebookFEConfig,
    *,
    stage_name: str,
    site_id: str,
    source_name: str | None = None,
    data_signature: str | None = None,
    config_signature: str | None = None,
) -> Path:
    root = ensure_dir(notebook_cfg.cache_root / stage_name / site_id)
    if source_name is not None:
        root = ensure_dir(root / source_name)
    if data_signature is not None and config_signature is not None:
        root = ensure_dir(root / f"{data_signature}__{config_signature}")
    return root


def write_json(path: str | Path, payload: dict[str, Any]) -> Path:
    path = Path(path)
    path.parent.mkdir(parents=True, exist_ok=True)
    path.write_text(json.dumps(payload, indent=2, default=str), encoding="utf-8")
    return path


def read_json(path: str | Path) -> dict[str, Any]:
    return json.loads(Path(path).read_text(encoding="utf-8"))


def write_stage_manifest(
    manifest_path: str | Path,
    *,
    stage_name: str,
    site_id: str,
    source_name: str | None,
    data_signature: str,
    config_signature: str,
    artifact_paths: dict[str, Any],
    success: bool,
    notes: list[str] | None = None,
) -> Path:
    payload = {
        "stage_name": stage_name,
        "site_id": site_id,
        "source_name": source_name,
        "data_signature": data_signature,
        "config_signature": config_signature,
        "artifact_paths": artifact_paths,
        "success": success,
        "notes": notes or [],
        "version": FEATURE_NOTEBOOK_VERSION,
    }
    return write_json(manifest_path, payload)


def has_valid_stage_cache(
    manifest_path: str | Path,
    *,
    expected_stage_name: str,
    expected_data_signature: str,
    expected_config_signature: str,
) -> bool:
    manifest_path = Path(manifest_path)
    if not manifest_path.exists():
        return False

    payload = read_json(manifest_path)
    return (
        payload.get("stage_name") == expected_stage_name
        and payload.get("data_signature") == expected_data_signature
        and payload.get("config_signature") == expected_config_signature
        and bool(payload.get("success"))
    )


NOTEBOOK_CONFIG_SIG = notebook_config_signature(fe_cfg)
print("Notebook config signature:", NOTEBOOK_CONFIG_SIG)
print("Cache root:", fe_cfg.cache_root)
```

```python
def read_raster_bundle(
    path: str | Path,
    *,
    site_id: str,
    source_name: str,
    band_names: list[str] | None = None,
) -> SourceRasterBundle:
    path = Path(path)

    with rasterio.open(path) as src:
        data = src.read().astype(np.float32)
        nodata = src.nodata

        if nodata is not None:
            data = np.where(data == nodata, np.nan, data)

        if band_names is None:
            band_names = [f"{source_name}_band_{i}" for i in range(1, src.count + 1)]

        if len(band_names) != src.count:
            raise ValueError(
                f"Band name count mismatch for {path}. "
                f"Expected {src.count}, got {len(band_names)}."
            )

        arrays = {name: data[i] for i, name in enumerate(band_names)}

        return SourceRasterBundle(
            site_id=site_id,
            source_name=source_name,
            arrays=arrays,
            transform=src.transform,
            crs=src.crs,
            nodata=nodata,
            metadata={
                "path": str(path),
                "count": src.count,
                "width": src.width,
                "height": src.height,
            },
        )


def grid_from_bundle(bundle: SourceRasterBundle) -> CanonicalGrid:
    h, w = bundle.shape
    return CanonicalGrid(
        width=w,
        height=h,
        transform=bundle.transform,
        crs=bundle.crs,
        source_name=bundle.source_name,
        nodata=np.nan,
    )


def reproject_band_to_grid(
    src_band: np.ndarray,
    *,
    src_transform: Affine,
    src_crs: Any,
    dst_grid: CanonicalGrid,
    resampling: Resampling = Resampling.bilinear,
    dst_nodata: float = np.nan,
) -> np.ndarray:
    dst = np.full((dst_grid.height, dst_grid.width), dst_nodata, dtype=np.float32)

    reproject(
        source=src_band.astype(np.float32),
        destination=dst,
        src_transform=src_transform,
        src_crs=src_crs,
        dst_transform=dst_grid.transform,
        dst_crs=dst_grid.crs,
        src_nodata=np.nan,
        dst_nodata=dst_nodata,
        resampling=resampling,
    )
    return dst


def align_bundle_to_grid(
    bundle: SourceRasterBundle,
    *,
    dst_grid: CanonicalGrid,
    resampling: Resampling,
) -> SourceRasterBundle:
    aligned = {}
    for band_name, band in bundle.arrays.items():
        aligned[band_name] = reproject_band_to_grid(
            band,
            src_transform=bundle.transform,
            src_crs=bundle.crs,
            dst_grid=dst_grid,
            resampling=resampling,
        )

    return SourceRasterBundle(
        site_id=bundle.site_id,
        source_name=bundle.source_name,
        arrays=aligned,
        transform=dst_grid.transform,
        crs=dst_grid.crs,
        nodata=np.nan,
        metadata={**bundle.metadata, "aligned_to": dst_grid.source_name},
    )


def summarize_array(name: str, arr: np.ndarray) -> dict[str, Any]:
    finite = np.isfinite(arr)
    return {
        "name": name,
        "shape": arr.shape,
        "finite_fraction": float(finite.mean()),
        "min": float(np.nanmin(arr)) if finite.any() else np.nan,
        "max": float(np.nanmax(arr)) if finite.any() else np.nan,
        "mean": float(np.nanmean(arr)) if finite.any() else np.nan,
        "std": float(np.nanstd(arr)) if finite.any() else np.nan,
    }
```

## Site assets and labeling bridges

```python
def feature_site_cache_root(site_id: str) -> Path:
    root = ensure_dir(fe_cfg.cache_root / "site_assets" / site_id)
    return root


def feature_site_naip_cache_root(site_id: str) -> Path:
    return ensure_dir(feature_site_cache_root(site_id) / "naip")


def feature_site_als_cache_root(site_id: str) -> Path:
    return ensure_dir(feature_site_cache_root(site_id) / "als_metadata")


def feature_site_3dep_cache_root(site_id: str) -> Path:
    return ensure_dir(feature_site_cache_root(site_id) / "3dep")


def feature_site_rap_cache_root(site_id: str) -> Path:
    return ensure_dir(feature_site_cache_root(site_id) / "rap")


def validate_cached_raster(path: str | Path) -> bool:
    path = Path(path)
    if not path.exists():
        return False

    try:
        with rasterio.open(path) as src:
            h = max(1, min(16, src.height))
            w = max(1, min(16, src.width))
            src.read([1], window=Window(0, 0, w, h))
        return True
    except Exception as e:
        LOGGER.warning("Cached raster validation failed for %s: %s", path, e)
        return False
```

```python
@dataclass
class SiteAssetBundle:
    site_id: str
    naip_path: Path | None = None
    als_metadata: list[dict[str, Any]] = field(default_factory=list)
    three_dep_path: Path | None = None
    rap_path: Path | None = None
    notes: list[str] = field(default_factory=list)


def prepare_naip_asset(site_id: str, *, force_refresh: bool = False) -> Path:
    site_base = site_to_remote_base(cfg, site_id)
    naip_name = site_to_tif_name(site_id)

    naip_local = feature_site_naip_cache_root(site_id) / naip_name
    naip_url = f"{site_base}/{cfg.data.naip_3dep_dir}/{naip_name}"

    use_cached = False
    if (not force_refresh) and naip_local.exists():
        if validate_cached_raster(naip_local):
            LOGGER.info("Using cached NAIP for site=%s: %s", site_id, naip_local)
            use_cached = True
        else:
            LOGGER.warning("Deleting corrupt cached NAIP for site=%s: %s", site_id, naip_local)
            naip_local.unlink(missing_ok=True)

    if not use_cached:
        LOGGER.info("Downloading NAIP for site=%s to %s", site_id, naip_local)
        download_file(naip_url, naip_local)
        if not validate_cached_raster(naip_local):
            raise RuntimeError(f"Downloaded NAIP for site={site_id} is unreadable: {naip_local}")

    return naip_local


def prepare_als_metadata(site_id: str, *, force_refresh: bool = False) -> list[dict[str, Any]]:
    als_cache_dir = feature_site_als_cache_root(site_id)
    als_meta_json = als_cache_dir / "als_metadata.json"

    if (not force_refresh) and als_meta_json.exists():
        LOGGER.info("Using cached ALS metadata for site=%s: %s", site_id, als_meta_json)
        return json.loads(als_meta_json.read_text(encoding="utf-8"))

    site_base = site_to_remote_base(cfg, site_id)
    als_url = f"{site_base}/{cfg.data.als_dir}"
    als_files = list_files_with_suffix(als_url, (".laz", ".las", ".copc.laz"))

    if not als_files:
        LOGGER.warning("No ALS files found for site=%s at %s", site_id, als_url)
        return []

    LOGGER.info("Downloading %d ALS file(s) for site=%s to extract metadata", len(als_files), site_id)

    records: list[dict[str, Any]] = []
    scratch_dir = Path(tempfile.mkdtemp(prefix=f"features_als_{site_id}_"))

    try:
        for entry in als_files:
            local_path = scratch_dir / entry["name"]
            download_file(entry["url"], local_path)

            meta = extract_als_metadata(local_path)
            meta["source_file"] = entry["name"]
            records.append(meta)

            try:
                local_path.unlink(missing_ok=True)
            except Exception as e:
                LOGGER.warning("Failed to remove temporary ALS file %s: %s", local_path, e)
    finally:
        try:
            scratch_dir.rmdir()
        except Exception:
            pass

    als_meta_json.write_text(json.dumps(records, indent=2), encoding="utf-8")
    LOGGER.info("Cached ALS metadata for site=%s at %s", site_id, als_meta_json)
    return records


def prepare_site_assets(site_id: str, *, force_refresh: bool = False) -> SiteAssetBundle:
    bundle = SiteAssetBundle(site_id=site_id)

    try:
        bundle.naip_path = prepare_naip_asset(site_id, force_refresh=force_refresh)
    except Exception as e:
        bundle.notes.append(f"NAIP unavailable: {e}")

    try:
        bundle.als_metadata = prepare_als_metadata(site_id, force_refresh=force_refresh)
    except Exception as e:
        bundle.notes.append(f"ALS metadata unavailable: {e}")

    # 3DEP and RAP are left as explicit placeholders for now.
    bundle.notes.append("3DEP adapter not yet connected in notebook runtime.")
    bundle.notes.append("RAP adapter not yet connected in notebook runtime.")

    return bundle
```

```python
def labeling_summary_dir() -> Path:
    return cfg.output.labeling_root / "summaries"


def labeling_manifest_dir() -> Path:
    return cfg.output.labeling_root / "manifests"


def load_labeling_objects_summary() -> pd.DataFrame:
    candidates = [
        labeling_summary_dir() / "objects_all.csv",
        cfg.output.labeling_root / "pipeline_runs",
    ]

    obj_csv = labeling_summary_dir() / "objects_all.csv"
    if obj_csv.exists():
        df = pd.read_csv(obj_csv)
        LOGGER.info("Loaded labeling object summary: %s (%d rows)", obj_csv, len(df))
        return df

    LOGGER.warning("No global labeling objects summary found at %s", obj_csv)
    return pd.DataFrame()


def load_labeling_artifacts_summary() -> pd.DataFrame:
    art_csv = labeling_summary_dir() / "artifacts_all.csv"
    if art_csv.exists():
        df = pd.read_csv(art_csv)
        LOGGER.info("Loaded labeling artifact summary: %s (%d rows)", art_csv, len(df))
        return df

    manifest_csv = labeling_manifest_dir() / "sprint4_artifacts.csv"
    if manifest_csv.exists():
        df = pd.read_csv(manifest_csv)
        LOGGER.info("Loaded labeling artifact manifest: %s (%d rows)", manifest_csv, len(df))
        return df

    LOGGER.warning("No labeling artifact summary/manifest found.")
    return pd.DataFrame()


LABEL_OBJECTS_DF = load_labeling_objects_summary()
LABEL_ARTIFACTS_DF = load_labeling_artifacts_summary()

print("LABEL_OBJECTS_DF:", LABEL_OBJECTS_DF.shape)
print("LABEL_ARTIFACTS_DF:", LABEL_ARTIFACTS_DF.shape)
```

```python
def label_objects_for_site(site_id: str) -> pd.DataFrame:
    if LABEL_OBJECTS_DF.empty:
        return pd.DataFrame()
    return LABEL_OBJECTS_DF[LABEL_OBJECTS_DF["site_id"] == site_id].copy()


def label_artifacts_for_site(site_id: str) -> pd.DataFrame:
    if LABEL_ARTIFACTS_DF.empty:
        return pd.DataFrame()
    return LABEL_ARTIFACTS_DF[LABEL_ARTIFACTS_DF["site_id"] == site_id].copy()


def best_label_artifact_for_site(site_id: str, resolution_m: float = 1.0) -> pd.Series | None:
    site_df = label_artifacts_for_site(site_id)
    if site_df.empty:
        return None

    if "resolution_m" in site_df.columns:
        site_df = site_df[np.isclose(site_df["resolution_m"].astype(float), float(resolution_m))].copy()

    if site_df.empty:
        return None

    sort_cols = [c for c in ["plot_id", "source_version"] if c in site_df.columns]
    if sort_cols:
        site_df = site_df.sort_values(sort_cols)

    return site_df.iloc[0]
```

# NAIP Grid Setup

```python
def build_canonical_grid_for_site(site_id: str, assets: SiteAssetBundle | None = None) -> CanonicalGrid:
    if assets is None:
        assets = prepare_site_assets(site_id, force_refresh=fe_cfg.force_refresh_assets)

    if assets.naip_path is None:
        raise ValueError(f"No NAIP path available for site={site_id}; cannot build canonical grid.")

    naip_bundle = read_raster_bundle(
        assets.naip_path,
        site_id=site_id,
        source_name="naip",
        band_names=None,
    )
    grid = grid_from_bundle(naip_bundle)
    LOGGER.info(
        "Built canonical grid for site=%s from %s | shape=(%d, %d) | pixel_size=%s",
        site_id,
        assets.naip_path.name,
        grid.height,
        grid.width,
        grid.pixel_size,
    )
    return grid
```

```python
def infer_naip_band_names(path: str | Path) -> list[str]:
    path = Path(path)
    with rasterio.open(path) as src:
        count = src.count

    if count == 4:
        return ["red", "green", "blue", "nir"]
    if count == 3:
        return ["red", "green", "blue"]
    return [f"band_{i}" for i in range(1, count + 1)]


def source_data_signature(
    *,
    site_id: str,
    source_name: str,
    asset_path: str | Path | None,
    extra_payload: dict[str, Any] | None = None,
) -> str:
    payload = {
        "site_id": site_id,
        "source_name": source_name,
        "asset_path": str(asset_path) if asset_path is not None else None,
        "extra": extra_payload or {},
    }
    return hash_payload(payload)


def source_config_signature(
    *,
    source_name: str,
    family_names: tuple[str, ...],
    notebook_cfg: NotebookFEConfig,
) -> str:
    payload = {
        "source_name": source_name,
        "family_names": family_names,
        "version": notebook_cfg.version,
        "resampling": notebook_cfg.resampling_by_source[source_name].name,
    }
    return hash_payload(payload)
```

```python
def load_naip_source_bundle(
    site_id: str,
    *,
    assets: SiteAssetBundle | None = None,
    align_to_canonical: bool = True,
    canonical_grid: CanonicalGrid | None = None,
    force_refresh: bool = False,
) -> SourceRasterBundle:
    if assets is None:
        assets = prepare_site_assets(site_id, force_refresh=force_refresh)

    if assets.naip_path is None:
        raise ValueError(f"NAIP is unavailable for site={site_id}")

    band_names = infer_naip_band_names(assets.naip_path)

    data_sig = source_data_signature(
        site_id=site_id,
        source_name="naip",
        asset_path=assets.naip_path,
        extra_payload={"band_names": band_names},
    )
    config_sig = source_config_signature(
        source_name="naip",
        family_names=fe_cfg.enabled_families["naip"],
        notebook_cfg=fe_cfg,
    )

    cache_dir = stage_cache_dir(
        fe_cfg,
        stage_name="source_bundle",
        site_id=site_id,
        source_name="naip",
        data_signature=data_sig,
        config_signature=config_sig,
    )
    manifest_path = cache_dir / "manifest.json"
    npz_path = cache_dir / "naip_bundle.npz"
    meta_path = cache_dir / "naip_bundle_meta.json"

    if (not force_refresh) and has_valid_stage_cache(
        manifest_path,
        expected_stage_name="source_bundle",
        expected_data_signature=data_sig,
        expected_config_signature=config_sig,
    ) and npz_path.exists() and meta_path.exists():
        meta = read_json(meta_path)
        arrays_raw = np.load(npz_path)
        arrays = {k: arrays_raw[k] for k in arrays_raw.files}

        bundle = SourceRasterBundle(
            site_id=site_id,
            source_name="naip",
            arrays=arrays,
            transform=Affine(*meta["transform"]),
            crs=meta["crs"],
            nodata=np.nan,
            metadata=meta["metadata"],
        )
        return bundle

    bundle = read_raster_bundle(
        assets.naip_path,
        site_id=site_id,
        source_name="naip",
        band_names=band_names,
    )

    if align_to_canonical:
        if canonical_grid is None:
            canonical_grid = build_canonical_grid_for_site(site_id, assets=assets)
        bundle = align_bundle_to_grid(
            bundle,
            dst_grid=canonical_grid,
            resampling=fe_cfg.resampling_by_source["naip"],
        )

    np.savez_compressed(npz_path, **bundle.arrays)
    write_json(
        meta_path,
        {
            "transform": list(bundle.transform)[:6],
            "crs": bundle.crs,
            "metadata": bundle.metadata,
        },
    )
    write_stage_manifest(
        manifest_path,
        stage_name="source_bundle",
        site_id=site_id,
        source_name="naip",
        data_signature=data_sig,
        config_signature=config_sig,
        artifact_paths={"npz_path": str(npz_path), "meta_path": str(meta_path)},
        success=True,
        notes=["Cached NAIP source bundle."],
    )
    return bundle
```

```python
def safe_divide(num: np.ndarray, denom: np.ndarray, fill_value: float = 0.0) -> np.ndarray:
    out = np.full_like(num, fill_value, dtype=np.float32)
    mask = np.isfinite(num) & np.isfinite(denom) & (np.abs(denom) > 1e-12)
    out[mask] = (num[mask] / denom[mask]).astype(np.float32)
    return out


def nan_to_num_copy(arr: np.ndarray, fill_value: float = 0.0) -> np.ndarray:
    out = arr.astype(np.float32, copy=True)
    out[~np.isfinite(out)] = fill_value
    return out


def gradient_magnitude(arr: np.ndarray) -> np.ndarray:
    arr = nan_to_num_copy(arr)
    gy, gx = np.gradient(arr)
    return np.sqrt(gx**2 + gy**2).astype(np.float32)


def local_std(arr: np.ndarray, size: int = 5) -> np.ndarray:
    arr = nan_to_num_copy(arr)
    mean = get_uniform_blur(arr, neighborhood_size=size)
    mean_sq = get_uniform_blur(arr**2, neighborhood_size=size)
    var = np.maximum(mean_sq - mean**2, 0.0)
    return np.sqrt(var).astype(np.float32)
```

```python
def run_naip_raw_family(bundle: SourceRasterBundle) -> FeatureFamilyResult:
    arrays = {f"naip_raw_{k}": v.astype(np.float32) for k, v in bundle.arrays.items()}
    return FeatureFamilyResult(
        site_id=bundle.site_id,
        source_name="naip",
        family_name="raw",
        arrays=arrays,
        metadata={"n_layers": len(arrays)},
    )


def run_naip_veg_idx_family(bundle: SourceRasterBundle) -> FeatureFamilyResult:
    bands = bundle.arrays
    out: dict[str, np.ndarray] = {}

    red = bands.get("red")
    green = bands.get("green")
    blue = bands.get("blue")
    nir = bands.get("nir")

    if red is not None and nir is not None:
        out["naip_idx_ndvi"] = safe_divide(nir - red, nir + red)

    if red is not None and green is not None and blue is not None:
        out["naip_idx_vari"] = safe_divide(green - red, green + red - blue)

    if red is not None and green is not None:
        out["naip_idx_g_minus_r"] = (green - red).astype(np.float32)

    if green is not None and blue is not None:
        out["naip_idx_g_minus_b"] = (green - blue).astype(np.float32)

    if not out:
        raise ValueError("NAIP veg_idx family requires at least RGB or NIR bands.")

    return FeatureFamilyResult(
        site_id=bundle.site_id,
        source_name="naip",
        family_name="veg_idx",
        arrays=out,
        metadata={"n_layers": len(out)},
    )


def run_naip_texture_family(
    bundle: SourceRasterBundle,
    *,
    entropy_window: int = 9,
    lbp_radius: int = 2,
    blur_sizes: tuple[int, ...] = (5, 11),
) -> FeatureFamilyResult:
    bands = bundle.arrays
    base = bands.get("nir", None)
    if base is None:
        base = bands.get("green", None)
    if base is None:
        base = next(iter(bands.values()))

    base_clean = nan_to_num_copy(base)

    out = {
        "naip_tex_gradmag": gradient_magnitude(base_clean),
        "naip_tex_localstd_5": local_std(base_clean, size=5),
        "naip_tex_entropy": get_entropy_feature(base_clean, neighborhood_size=entropy_window).astype(np.float32),
        "naip_tex_lbp_r2": get_fast_lbp_texture(base_clean, radius=lbp_radius).astype(np.float32),
    }

    for size in blur_sizes:
        out[f"naip_tex_blur_{size}"] = get_uniform_blur(base_clean, neighborhood_size=size).astype(np.float32)

    return FeatureFamilyResult(
        site_id=bundle.site_id,
        source_name="naip",
        family_name="texture",
        arrays=out,
        metadata={
            "base_band": "nir" if "nir" in bands else ("green" if "green" in bands else "first_band"),
            "n_layers": len(out),
        },
    )


def run_naip_multiscale_family(
    bundle: SourceRasterBundle,
    *,
    neighborhood_sizes: tuple[int, ...] = (3, 7, 15),
) -> FeatureFamilyResult:
    bands = bundle.arrays
    out: dict[str, np.ndarray] = {}

    selected_names = [name for name in ["red", "green", "nir"] if name in bands]
    if not selected_names:
        selected_names = list(bands.keys())[:1]

    for band_name in selected_names:
        band = nan_to_num_copy(bands[band_name])
        for size in neighborhood_sizes:
            out[f"naip_ms_mean_{band_name}_{size}"] = get_uniform_blur(band, neighborhood_size=size).astype(np.float32)
            out[f"naip_ms_std_{band_name}_{size}"] = local_std(band, size=size)

    return FeatureFamilyResult(
        site_id=bundle.site_id,
        source_name="naip",
        family_name="multiscale",
        arrays=out,
        metadata={"selected_bands": selected_names, "n_layers": len(out)},
    )
```

```python
NAIP_FAMILY_RUNNERS = {
    "raw": run_naip_raw_family,
    "veg_idx": run_naip_veg_idx_family,
    "texture": run_naip_texture_family,
    "multiscale": run_naip_multiscale_family,
}


def flatten_family_results(results: list[FeatureFamilyResult]) -> dict[str, np.ndarray]:
    merged: dict[str, np.ndarray] = {}
    for result in results:
        overlap = set(merged).intersection(result.arrays)
        if overlap:
            raise ValueError(f"Duplicate feature names detected: {sorted(overlap)}")
        merged.update(result.arrays)
    return merged


def run_naip_feature_families(
    site_id: str,
    *,
    assets: SiteAssetBundle | None = None,
    canonical_grid: CanonicalGrid | None = None,
    family_names: tuple[str, ...] | None = None,
    force_refresh: bool = False,
) -> tuple[SourceRasterBundle, list[FeatureFamilyResult], dict[str, np.ndarray]]:
    if assets is None:
        assets = prepare_site_assets(site_id, force_refresh=force_refresh)
    if canonical_grid is None:
        canonical_grid = build_canonical_grid_for_site(site_id, assets=assets)

    bundle = load_naip_source_bundle(
        site_id,
        assets=assets,
        align_to_canonical=True,
        canonical_grid=canonical_grid,
        force_refresh=force_refresh,
    )

    family_names = family_names or fe_cfg.enabled_families["naip"]

    results: list[FeatureFamilyResult] = []
    for family_name in family_names:
        runner = NAIP_FAMILY_RUNNERS[family_name]
        result = runner(bundle)
        results.append(result)

    merged = flatten_family_results(results)
    LOGGER.info(
        "Ran NAIP families for site=%s | families=%s | total_layers=%d",
        site_id,
        family_names,
        len(merged),
    )
    return bundle, results, merged
```

```python
def feature_stack_frame(layer_dict: dict[str, np.ndarray]) -> pd.DataFrame:
    rows = [summarize_array(name, arr) for name, arr in layer_dict.items()]
    return pd.DataFrame(rows).sort_values("name").reset_index(drop=True)


def preview_naip_run(site_id: str):
    assets = prepare_site_assets(site_id, force_refresh=fe_cfg.force_refresh_assets)
    grid = build_canonical_grid_for_site(site_id, assets=assets)
    bundle, family_results, merged = run_naip_feature_families(
        site_id,
        assets=assets,
        canonical_grid=grid,
        force_refresh=fe_cfg.force_refresh_features,
    )

    print("Site:", site_id)
    print("NAIP path:", assets.naip_path)
    print("Grid shape:", (grid.height, grid.width))
    print("Grid pixel size:", grid.pixel_size)
    print("Raw band names:", list(bundle.arrays.keys()))
    print("Families run:", [r.family_name for r in family_results])
    print("Total merged layers:", len(merged))

    return {
        "assets": assets,
        "grid": grid,
        "bundle": bundle,
        "family_results": family_results,
        "merged": merged,
        "summary": feature_stack_frame(merged),
    }
```

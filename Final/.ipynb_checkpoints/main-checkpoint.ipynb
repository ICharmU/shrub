```python
#progress_bar = tqdm(total=len(trials), desc="Resolved trials")
trial_bar = tqdm(total=len(trials), desc="Resolved trials", position=0)
unit_bar = tqdm(total=0, desc="Completed work units", position=1)

last_completed = {"trial_id": None, "unit_id": None, "stage_name": None}
worker_history = []
iteration_idx = 0

def render_worker_dashboard(
    *,
    trials,
    scheduler,
    active_job=None,
    candidates=None,
    tail_log_path=None,
    already_refreshed: bool = False,
    header_text: str | None = None,
):
    clear_output(wait=True)

    if already_refreshed:
        refreshed_trials = list(trials)
    else:
        refreshed_trials = [controller.load_trial(t.trial_id) for t in trials]

    resolved = sum(t.status in {"success", "failed"} for t in refreshed_trials)

    total_units = 0
    complete_units = 0
    for t in refreshed_trials:
        res = t.resolution or {}
        total_units += int(res.get("total_units", 0))
        complete_units += int(res.get("complete_units", 0))

    trial_bar.total = max(len(refreshed_trials), 1)
    trial_bar.n = resolved
    trial_bar.refresh()

    unit_bar.total = max(total_units, 1)
    unit_bar.n = complete_units
    unit_bar.refresh()

    if header_text:
        print(header_text)

    snapshot_df = scheduler.scheduler_snapshot_frame(trials=refreshed_trials)
    if not snapshot_df.empty:
        display(snapshot_df.sort_values(["status", "trial_id"]).reset_index(drop=True))

    if candidates:
        cand_df = scheduler.candidate_frame(candidates=candidates, top_n=8)
        if not cand_df.empty:
            display(cand_df)

    if active_job is not None:
        trial, pipeline, unit = active_job

        latest_trial = None
        for t in refreshed_trials:
            if t.trial_id == trial.trial_id:
                latest_trial = t
                break
        if latest_trial is None:
            latest_trial = controller.load_trial(trial.trial_id)

        current_trial_df = scheduler.current_trial_frame(trial=latest_trial)
        if not current_trial_df.empty:
            display(current_trial_df)

        display(pd.DataFrame([{
            "active_trial": trial.trial_id,
            "pipeline": pipeline.pipeline_name,
            "unit_id": unit.get("unit_id"),
            "stage_name": unit.get("stage_name"),
            "scope": unit.get("scope"),
            "site_id": unit.get("site_id"),
            "plot_id": unit.get("plot_id"),
            "source_version": unit.get("source_version"),
            "priority": unit.get("priority"),
        }]))

    if tail_log_path is not None:
        print(tail_text_file(tail_log_path, n_lines=25))

    return refreshed_trials

log_path = cfg.output.logs_root / "pipeline.log"

active_trials = [controller.load_trial(t.trial_id) for t in trials]

idle_polls = 0
max_idle_polls = 20
idle_sleep_sec = 15.0

while True:
    active_trials = [controller.load_trial(t.trial_id) for t in trials]

    # Show current known state BEFORE expensive refresh/enumeration.
    render_worker_dashboard(
        trials=active_trials,
        scheduler=scheduler,
        active_job=None,
        candidates=None,
        tail_log_path=log_path,
        already_refreshed=True,
        header_text="Collecting runnable candidates...",
    )

    force_refresh = (idle_polls == 0) or (iteration_idx == 0)
    candidates = scheduler.collect_candidate_jobs(
        trials=active_trials,
        pipelines={"labeling": labeling_pipeline},
        force_refresh=force_refresh,
    )

    # Refresh trial objects after candidate collection, since collect_candidate_jobs
    # may have updated work units/resolution on disk.
    active_trials = [controller.load_trial(t.trial_id) for t in trials]

    trial_resolution_df = pd.DataFrame([
        {
            "trial_id": t.trial_id,
            "status": t.status,
            **(t.resolution or {}),
            "n_work_units_dict": len(t.work_units or {}),
        }
        for t in active_trials
    ])
    
    display(trial_resolution_df)
    print("n_candidates:", len(candidates))

    if not candidates:
        idle_polls += 1

        active_trials = render_worker_dashboard(
            trials=active_trials,
            scheduler=scheduler,
            active_job=None,
            candidates=[],
            tail_log_path=log_path,
            already_refreshed=True,
            header_text=f"No runnable jobs found. Idle poll {idle_polls}/{max_idle_polls}",
        )

        if idle_polls >= max_idle_polls:
            break

        time.sleep(idle_sleep_sec)
        continue

    idle_polls = 0
    top = candidates[0]
    trial, pipeline, unit = top["trial"], top["pipeline"], top["unit"]

    active_trials = render_worker_dashboard(
        trials=active_trials,
        scheduler=scheduler,
        active_job=(trial, pipeline, unit),
        candidates=candidates,
        tail_log_path=log_path,
        already_refreshed=True,
        header_text="Selected next runnable job",
    )

    claimed, payload = scheduler.claim_unit(trial=trial, pipeline=pipeline, unit=unit)
    if not claimed:
        continue

    scheduler.run_claimed_job(
        trial=trial,
        pipeline=pipeline,
        unit=unit,
        state=state,
    )

    controller.save_state(state)

    iteration_idx += 1
    latest_trial = controller.load_trial(trial.trial_id)
    worker_history.append(
        {
            "iteration": iteration_idx,
            "trial_id": trial.trial_id,
            "pipeline_name": pipeline.pipeline_name,
            "unit_id": unit.get("unit_id"),
            "stage_name": unit.get("stage_name"),
            "scope": unit.get("scope"),
            "site_id": unit.get("site_id"),
            "plot_id": unit.get("plot_id"),
            "source_version": unit.get("source_version"),
            "trial_status_after": latest_trial.status,
            "resolution_after": dict(latest_trial.resolution or {}),
        }
    )

trial_bar.close()
unit_bar.close()
```

```python
def validate_transfer_artifacts_for_trial(trial, pipeline):
    latest = controller.load_trial(trial.trial_id)
    rows = []

    for unit in (latest.work_units or {}).values():
        if unit.get("pipeline_name") != "labeling":
            continue
        if unit.get("stage_name") != "transfer":
            continue

        site_id = unit.get("site_id")
        plot_id = unit.get("plot_id")
        source_version = unit.get("source_version")
        status = unit.get("status")

        paths = pipeline.artifact_paths_for_plot(site_id, plot_id, source_version)
        required = {
            "binary_path": paths["binary_path"].exists(),
            "confidence_path": paths["confidence_path"].exists(),
            "object_id_path": paths["object_id_path"].exists(),
            "object_table_path": paths["object_table_path"].exists(),
            "qa_path": paths["qa_path"].exists(),
        }

        rows.append({
            "trial_id": latest.trial_id,
            "site_id": site_id,
            "plot_id": plot_id,
            "source_version": source_version,
            "unit_status": status,
            **required,
            "all_required_exist": all(required.values()),
        })

    return pd.DataFrame(rows)

transfer_validation_df = pd.concat(
    [
        validate_transfer_artifacts_for_trial(trial, pipelines_by_trial[trial.trial_id]["labeling"])
        for trial in trials
    ],
    ignore_index=True,
)

display(transfer_validation_df)
display(transfer_validation_df[transfer_validation_df["all_required_exist"] == False])
```

```python
feature_trial_bar = tqdm(total=len(feature_trials), desc="Resolved feature trials", position=0)
feature_unit_bar = tqdm(total=0, desc="Completed feature work units", position=1)

feature_worker_history = []
feature_iteration_idx = 0

feature_log_path = cfg.output.logs_root / "pipeline.log"

def render_feature_worker_dashboard(
    *,
    trials,
    scheduler,
    active_job=None,
    candidates=None,
    tail_log_path=None,
    already_refreshed: bool = False,
    header_text: str | None = None,
):
    clear_output(wait=True)

    if already_refreshed:
        refreshed_trials = list(trials)
    else:
        refreshed_trials = [controller.load_trial(t.trial_id) for t in trials]

    resolved = sum(t.status in {"success", "failed"} for t in refreshed_trials)

    total_units = 0
    complete_units = 0
    for t in refreshed_trials:
        res = t.resolution or {}
        total_units += int(res.get("total_units", 0))
        complete_units += int(res.get("complete_units", 0))

    feature_trial_bar.total = max(len(refreshed_trials), 1)
    feature_trial_bar.n = resolved
    feature_trial_bar.refresh()

    feature_unit_bar.total = max(total_units, 1)
    feature_unit_bar.n = complete_units
    feature_unit_bar.refresh()

    if header_text:
        print(header_text)

    snapshot_df = scheduler.scheduler_snapshot_frame(trials=refreshed_trials)
    if not snapshot_df.empty:
        display(snapshot_df.sort_values(["status", "trial_id"]).reset_index(drop=True))

    if candidates:
        cand_df = scheduler.candidate_frame(candidates=candidates, top_n=8)
        if not cand_df.empty:
            display(cand_df)

    if active_job is not None:
        trial, pipeline, unit = active_job

        latest_trial = None
        for t in refreshed_trials:
            if t.trial_id == trial.trial_id:
                latest_trial = t
                break
        if latest_trial is None:
            latest_trial = controller.load_trial(trial.trial_id)

        current_trial_df = scheduler.current_trial_frame(trial=latest_trial)
        if not current_trial_df.empty:
            display(current_trial_df)

        display(pd.DataFrame([{
            "active_trial": trial.trial_id,
            "pipeline": pipeline.pipeline_name,
            "unit_id": unit.get("unit_id"),
            "stage_name": unit.get("stage_name"),
            "scope": unit.get("scope"),
            "site_id": unit.get("site_id"),
            "plot_id": unit.get("plot_id"),
            "source_version": unit.get("source_version"),
            "priority": unit.get("priority"),
        }]))

    if tail_log_path is not None:
        print(tail_text_file(tail_log_path, n_lines=25))

    return refreshed_trials

feature_active_trials = [controller.load_trial(t.trial_id) for t in feature_trials]

feature_idle_polls = 0
feature_max_idle_polls = 20
feature_idle_sleep_sec = 15.0

while True:
    feature_active_trials = [controller.load_trial(t.trial_id) for t in feature_trials]

    render_feature_worker_dashboard(
        trials=feature_active_trials,
        scheduler=feature_scheduler,
        active_job=None,
        candidates=None,
        tail_log_path=feature_log_path,
        already_refreshed=True,
        header_text="Collecting runnable FE candidates...",
    )

    force_refresh = (feature_idle_polls == 0) or (feature_iteration_idx == 0)

    candidates = feature_scheduler.collect_candidate_jobs(
        trials=feature_active_trials,
        pipelines=feature_pipelines_by_trial,
        force_refresh=force_refresh,
    )

    feature_active_trials = [controller.load_trial(t.trial_id) for t in feature_trials]

    feature_trial_resolution_df = pd.DataFrame([
        {
            "trial_id": t.trial_id,
            "status": t.status,
            **(t.resolution or {}),
            "n_work_units_dict": len(t.work_units or {}),
        }
        for t in feature_active_trials
    ])

    display(feature_trial_resolution_df)
    print("n_candidates:", len(candidates))

    if not candidates:
        feature_idle_polls += 1

        feature_active_trials = render_feature_worker_dashboard(
            trials=feature_active_trials,
            scheduler=feature_scheduler,
            active_job=None,
            candidates=[],
            tail_log_path=feature_log_path,
            already_refreshed=True,
            header_text=f"No runnable FE jobs found. Idle poll {feature_idle_polls}/{feature_max_idle_polls}",
        )

        if feature_idle_polls >= feature_max_idle_polls:
            break

        time.sleep(feature_idle_sleep_sec)
        continue

    feature_idle_polls = 0
    top = candidates[0]
    trial, pipeline, unit = top["trial"], top["pipeline"], top["unit"]

    feature_active_trials = render_feature_worker_dashboard(
        trials=feature_active_trials,
        scheduler=feature_scheduler,
        active_job=(trial, pipeline, unit),
        candidates=candidates,
        tail_log_path=feature_log_path,
        already_refreshed=True,
        header_text="Selected next runnable FE job",
    )

    claimed, payload = feature_scheduler.claim_unit(
        trial=trial,
        pipeline=pipeline,
        unit=unit,
    )
    if not claimed:
        continue

    feature_scheduler.run_claimed_job(
        trial=trial,
        pipeline=pipeline,
        unit=unit,
        state=state,
    )

    controller.save_state(state)

    feature_iteration_idx += 1
    latest_trial = controller.load_trial(trial.trial_id)

    feature_worker_history.append(
        {
            "iteration": feature_iteration_idx,
            "trial_id": trial.trial_id,
            "pipeline_name": pipeline.pipeline_name,
            "unit_id": unit.get("unit_id"),
            "stage_name": unit.get("stage_name"),
            "scope": unit.get("scope"),
            "site_id": unit.get("site_id"),
            "plot_id": unit.get("plot_id"),
            "source_version": unit.get("source_version"),
            "trial_status_after": latest_trial.status,
            "resolution_after": dict(latest_trial.resolution or {}),
        }
    )

feature_trial_bar.close()
feature_unit_bar.close()
```

```python
# source/bridge imports already present in your notebook ecosystem
from Final.features.canonical_grid import build_canonical_grid_for_site
from Final.features.chunking import build_chunk_manifest
from Final.features.labeling_bridge import (
    load_labeling_artifacts_summary,
    load_labeling_objects_summary,
    label_objects_for_site,
)
from Final.features.object_aggregation import build_and_persist_object_feature_table
from Final.features.source_naip import prepare_naip_asset, run_naip_chunked_pipeline
from Final.features.source_als import prepare_als_metadata, run_als_chunked_pipeline
from Final.features.source_3dep import prepare_3dep_asset, run_3dep_chunked_pipeline
from Final.features.source_rap import prepare_rap_asset, run_rap_chunked_pipeline
from Final.features.assets import find_cached_rap_asset, find_cached_3dep_asset

PIPELINE_OBJECT_CACHE = {}

def make_feature_ops():
    return FeaturePipelineOps(
        prepare_site_assets=prepare_site_assets,  # your existing notebook/site-assets helper
        build_canonical_grid_for_site=build_canonical_grid_for_site,
        build_chunk_manifest=build_chunk_manifest,
        ensure_source_ready=ensure_source_ready,  # your existing notebook helper
        prepare_naip_asset=prepare_naip_asset,
        prepare_als_metadata=prepare_als_metadata,
        prepare_3dep_asset=prepare_3dep_asset,
        prepare_rap_asset=prepare_rap_asset,
        find_cached_rap_asset=find_cached_rap_asset,
        find_cached_3dep_asset=find_cached_3dep_asset,
        run_naip_chunked_pipeline=run_naip_chunked_pipeline,
        run_als_chunked_pipeline=run_als_chunked_pipeline,
        run_3dep_chunked_pipeline=run_3dep_chunked_pipeline,
        run_rap_chunked_pipeline=run_rap_chunked_pipeline,
        label_objects_for_site=label_objects_for_site,
        build_and_persist_object_feature_table=build_and_persist_object_feature_table,
    )

from Final.artifact_store import HybridArtifactStore
from Final.features.pipeline import FeaturePipeline, FeaturePipelineConfig, FeaturePipelineOps
from Final.labeling.pipeline import LabelingPipeline, LabelingPipelineConfig

PIPELINE_OBJECT_CACHE = {}

def materialize_feature_pipeline(*, cfg_local, feature_signature=None):
    key = ("features", feature_signature or "active")
    if key in PIPELINE_OBJECT_CACHE:
        return PIPELINE_OBJECT_CACHE[key]

    pipe = FeaturePipeline(
        cfg_local,
        ops=make_feature_ops(),
        pipeline_config=FeaturePipelineConfig(
            canonical_grid_source=cfg_local.features.canonical_grid_source
        ),
    )

    print("materialized feature artifact_store:", type(pipe.artifact_store).__name__)
    if not isinstance(pipe.artifact_store, HybridArtifactStore):
        raise RuntimeError(
            "FeaturePipeline did not materialize with HybridArtifactStore. "
            "Refusing to continue because the selected FE registry uses pruned/staged chunk paths."
        )

    PIPELINE_OBJECT_CACHE[key] = pipe
    return pipe

def materialize_labeling_pipeline(*, cfg_local):
    key = ("labeling", "active")
    if key in PIPELINE_OBJECT_CACHE:
        return PIPELINE_OBJECT_CACHE[key]

    pipe = LabelingPipeline(
        cfg_local,
        pipeline_config=LabelingPipelineConfig(),
    )

    print("materialized labeling artifact_store:", type(pipe.artifact_store).__name__)
    PIPELINE_OBJECT_CACHE[key] = pipe
    return pipe

def materialize_downstream_support(*, feature_signature=None, cfg_override=None):
    cfg_local = cfg_override or make_downstream_cfg()
    feature_pipe = materialize_feature_pipeline(
        cfg_local=cfg_local,
        feature_signature=feature_signature,
    )
    labeling_pipe = materialize_labeling_pipeline(
        cfg_local=cfg_local,
    )
    return cfg_local, feature_pipe, labeling_pipe
```

```python
import hashlib
import json
import numpy as np
from pathlib import Path

ASSEMBLED_LAYER_CACHE = {}

def stable_sig(payload: dict) -> str:
    return hashlib.sha256(
        json.dumps(payload, sort_keys=True, default=str).encode("utf-8")
    ).hexdigest()[:16]

def assembled_bundle_cache_path(site_id, feature_signature, feature_profile, resolution_m):
    cache_root = Path(cfg.output.features_root) / "notebook_cache" / "assembled_site_bundles"
    cache_root.mkdir(parents=True, exist_ok=True)

    sig = stable_sig({
        "site_id": site_id,
        "feature_signature": feature_signature,
        "feature_profile": feature_profile,
        "resolution_m": resolution_m,
    })
    return cache_root / f"{site_id}__{sig}.npz"

def load_cached_assembled_bundle(path: Path):
    if not path.exists():
        return None
    with np.load(path, allow_pickle=False) as data:
        return {k: data[k] for k in data.files}

def save_cached_assembled_bundle(path: Path, arrays: dict[str, np.ndarray]):
    path.parent.mkdir(parents=True, exist_ok=True)
    np.savez_compressed(path, **arrays)
    return path
```

```python
# Shared helpers: feature selection, label hydration, crop/alignment, and FE-ingestion preview

_SITE_PATCH_CACHE = {}

def make_section_cfg_for_sites(base_cfg, sites):
    local_cfg = copy.deepcopy(base_cfg)
    local_cfg.site.sites = list(sites)
    local_cfg.resolve()
    return local_cfg

def resolve_feature_pipeline(feature_signature=None):
    feature_signature = feature_signature or REAL_FEATURE_SIG

    if feature_signature in FEATURE_PIPELINES_BY_SIG:
        pipe = FEATURE_PIPELINES_BY_SIG[feature_signature]
        return pipe

    cfg_local, pipe, _ = materialize_downstream_support(feature_signature=feature_signature)
    FEATURE_PIPELINES_BY_SIG[feature_signature] = pipe
    return pipe

def resolve_labeling_pipeline():
    _, _, labeling_pipe = materialize_downstream_support(feature_signature=REAL_FEATURE_SIG)
    return labeling_pipe

def choose_pixel_layer_names(registry, feature_profile: str = "all_sources") -> list[str]:
    df = stack_registry_frame(registry)
    if df.empty:
        return []

    work = df.copy()
    work["base_layer_name"] = work["layer_name"].astype(str).str.split("::").str[0]

    profile_to_sources = {
        "naip": {"naip"},
        "naip_als": {"naip", "als"},
        "naip_als_3dep": {"naip", "als", "3dep"},
        "all_sources": {"naip", "als", "3dep", "rap"},
    }

    wanted_sources = profile_to_sources.get(feature_profile)
    if wanted_sources is None:
        raise ValueError(f"Unsupported feature_profile={feature_profile}")

    # Prefer the explicit registry metadata, not layer-name prefixes.
    if "source_name" in work.columns:
        work = work[work["source_name"].isin(wanted_sources)].copy()
    else:
        # conservative fallback only if source_name metadata is absent
        prefixes = tuple(f"{src}_" for src in wanted_sources if src != "3dep")
        mask = work["base_layer_name"].str.startswith(prefixes)
        # keep terrain-like 3DEP fallback if requested
        if "3dep" in wanted_sources:
            terrain_tokens = ("terrain", "elevation", "slope", "aspect", "northness", "eastness", "curvature", "ruggedness", "tpi", "hillshade")
            mask = mask | work["base_layer_name"].str.contains("|".join(terrain_tokens), case=False, regex=True)
        work = work[mask].copy()

    selected = sorted(work["base_layer_name"].dropna().unique().tolist())
    return selected

def choose_existing_label_artifact(site_id: str, resolution_m: float = 1.0) -> pd.Series | None:
    site_df = label_artifacts_for_site(site_id)
    if site_df.empty:
        print(f"[choose_existing_label_artifact] no label artifacts for site={site_id}")
        return None

    work = site_df.copy()

    if "resolution_m" in work.columns:
        work = work[np.isclose(work["resolution_m"].astype(float), float(resolution_m))].copy()

    if work.empty:
        print(f"[choose_existing_label_artifact] no artifacts at resolution={resolution_m} for site={site_id}")
        return None

    def _exists(p):
        if p is None or (isinstance(p, float) and np.isnan(p)):
            return False
        try:
            return Path(p).exists()
        except Exception:
            return False

    work["binary_exists"] = work["binary_mask_path"].map(_exists) if "binary_mask_path" in work.columns else False
    work["confidence_exists"] = work["confidence_mask_path"].map(_exists) if "confidence_mask_path" in work.columns else False

    # Prefer revised-like variants, then existing binary paths, then confidence paths
    if "source_version" in work.columns:
        work["source_version_str"] = work["source_version"].astype(str).str.lower()
    else:
        work["source_version_str"] = ""

    if "label_variant" in work.columns:
        work["label_variant_str"] = work["label_variant"].astype(str).str.lower()
    else:
        work["label_variant_str"] = ""

    work["prefer_revised"] = (
        work["source_version_str"].str.contains("revised", na=False)
        | work["label_variant_str"].str.contains("revised", na=False)
    )

    sort_cols = []
    ascending = []

    # revised first
    sort_cols.append("prefer_revised")
    ascending.append(False)

    # existing binary path first
    sort_cols.append("binary_exists")
    ascending.append(False)

    # then confidence path existence
    sort_cols.append("confidence_exists")
    ascending.append(False)

    # stable tiebreakers if present
    for c in ["plot_id", "source_version", "label_variant"]:
        if c in work.columns:
            sort_cols.append(c)
            ascending.append(True)

    work = work.sort_values(sort_cols, ascending=ascending).reset_index(drop=True)

    print("[choose_existing_label_artifact] candidates:")
    keep_cols = [c for c in [
        "site_id", "plot_id", "source_version", "label_variant", "resolution_m",
        "binary_mask_path", "confidence_mask_path", "binary_exists", "confidence_exists"
    ] if c in work.columns]
    display(work[keep_cols].head(10))

    # prefer an existing binary path
    existing = work[work["binary_exists"]].copy()
    if not existing.empty:
        return existing.iloc[0]

    # fallback to first row for debugging visibility
    return work.iloc[0]

def find_live_labeling_pipeline():
    from Final.labeling.pipeline import LabelingPipeline

    candidates = []
    for name, obj in globals().items():
        try:
            if isinstance(obj, LabelingPipeline):
                candidates.append((name, obj))
        except Exception:
            pass

    if not candidates:
        raise RuntimeError("No live LabelingPipeline object found in globals().")

    preferred = ["labeling_pipeline", "label_pipe", "pipeline", "ACTIVE_LABELING_PIPELINE"]
    by_name = dict(candidates)
    for name in preferred:
        if name in by_name:
            return by_name[name]
    return candidates[0][1]


def hydrate_label_artifact_if_needed(path_str: str | None, *, labeling_pipe):
    if not path_str:
        return None

    p = Path(path_str)
    if p.exists():
        return p

    # Try to infer rel_path from the absolute path
    # We look for the "labeling/..." suffix inside the local path.
    parts = p.parts
    try:
        idx = parts.index("labeling")
        rel_path = "/".join(parts[idx:])
    except ValueError:
        rel_path = None

    if rel_path is not None:
        try:
            if labeling_pipe.artifact_store.exists(rel_path):
                target_local = p
                target_local.parent.mkdir(parents=True, exist_ok=True)
                pulled = labeling_pipe.artifact_store.pull(rel_path, local_path=target_local)
                return Path(pulled)
        except Exception as e:
            print(f"[hydrate_label_artifact_if_needed] remote hydrate failed for {rel_path}: {e}")

    return None


def hydrate_label_artifact_if_needed(path_str: str | None, *, labeling_pipe):
    if not path_str:
        return None

    p = Path(path_str)
    if p.exists():
        return p

    parts = p.parts
    rel_path = None
    try:
        idx = parts.index("labeling")
        rel_path = "/".join(parts[idx:])
    except ValueError:
        pass

    if rel_path is not None:
        try:
            if labeling_pipe.artifact_store.exists(rel_path):
                p.parent.mkdir(parents=True, exist_ok=True)
                pulled = labeling_pipe.artifact_store.pull(rel_path, local_path=p)
                return Path(pulled)
        except Exception as e:
            print(f"[label hydrate] failed for {rel_path}: {e}")

    return None

def read_label_bundle_for_site(site_id: str, resolution_m: float = 1.0, labeling_pipe=None):
    labeling_pipe = labeling_pipe or resolve_labeling_pipeline()

    art = choose_existing_label_artifact(site_id, resolution_m=resolution_m)
    if art is None:
        raise FileNotFoundError(f"No label artifacts found for site={site_id} at resolution={resolution_m}")

    binary_path = hydrate_label_artifact_if_needed(
        art.get("binary_mask_path"),
        labeling_pipe=labeling_pipe,
    )
    confidence_path = hydrate_label_artifact_if_needed(
        art.get("confidence_mask_path"),
        labeling_pipe=labeling_pipe,
    )

    if binary_path is None or not binary_path.exists():
        raise RuntimeError(
            f"No usable binary label raster found for site={site_id}. "
            f"Chosen path={art.get('binary_mask_path')}"
        )

    with rasterio.open(binary_path) as src:
        mask = src.read(1)
        transform = src.transform
        crs = src.crs

    confidence = None
    if confidence_path is not None and confidence_path.exists():
        with rasterio.open(confidence_path) as src:
            confidence = src.read(1)

    return {
        "site_id": site_id,
        "artifact_row": art,
        "binary_path": binary_path,
        "confidence_path": confidence_path if confidence_path is not None and confidence_path.exists() else None,
        "mask": mask,
        "confidence": confidence,
        "transform": transform,
        "crs": crs,
    }

def _clip_window_for_bounds(canonical_grid, bounds):
    win = from_bounds(bounds.left, bounds.bottom, bounds.right, bounds.top, transform=canonical_grid.transform)
    row_off = max(0, int(np.floor(win.row_off)))
    col_off = max(0, int(np.floor(win.col_off)))
    height = min(canonical_grid.height - row_off, int(np.ceil(win.height)))
    width = min(canonical_grid.width - col_off, int(np.ceil(win.width)))
    return row_off, col_off, height, width

def _match_shape(arr2d: np.ndarray, target_shape: tuple[int, int], order: int = 1) -> np.ndarray:
    if tuple(arr2d.shape) == tuple(target_shape):
        return arr2d.astype(np.float32, copy=False)
    resized = resize(
        arr2d,
        target_shape,
        order=order,
        preserve_range=True,
        anti_aliasing=(order != 0),
    )
    return resized.astype(np.float32)

_SITE_PATCH_CACHE = {}

from Final.features.models import RasterStackRegistry, RasterLayerRecord

def registry_from_df(reg_df: pd.DataFrame, site_id: str, config_signature: str) -> RasterStackRegistry:
    if reg_df.empty:
        return RasterStackRegistry(site_id=site_id, config_signature=config_signature, layers=[])

    layers = []
    for _, row in reg_df.iterrows():
        payload = row.to_dict()

        # keep only fields RasterLayerRecord understands
        allowed = {
            "site_id", "source_name", "family_name", "layer_name",
            "chunk_id", "rel_path", "local_path", "remote_ref",
            "storage_tier", "chunked", "dtype", "shape", "notes",
        }
        payload = {k: v for k, v in payload.items() if k in allowed}

        # normalize shape
        if "shape" in payload and isinstance(payload["shape"], list):
            payload["shape"] = tuple(payload["shape"])

        layers.append(RasterLayerRecord(**payload))

    return RasterStackRegistry(
        site_id=site_id,
        config_signature=config_signature,
        layers=layers,
    )

def materialize_site_feature_label_patch(
    site_id: str,
    *,
    feature_signature: str,
    feature_profile: str,
    resolution_m: float = 1.0,
    force_refresh: bool = False,
    registry_path: str | Path | None = None,
    cfg=None,
):
    cfg = cfg or default_config()

    cache_key = (site_id, feature_signature, feature_profile, resolution_m, str(registry_path))
    if (not force_refresh) and cache_key in _SITE_PATCH_CACHE:
        return _SITE_PATCH_CACHE[cache_key]

    cfg_local, feature_pipe, labeling_pipe = materialize_downstream_support(
        feature_signature=feature_signature,
        cfg_override=cfg if cfg is not None else None,
    )

    site_assets = feature_pipe.prepare_site_assets_for_pipeline(site_id, force_refresh=force_refresh)
    canonical_grid = feature_pipe.ops.build_canonical_grid_for_site(
        site_id,
        artifact_store=feature_pipe.artifact_store,
        assets=site_assets,
        force_refresh=False,
    )
    chunk_manifest = feature_pipe.ops.build_chunk_manifest(
        canonical_grid,
        artifact_store=feature_pipe.artifact_store,
        force_refresh=False,
    )

    if registry_path is not None:
        reg_df = load_registry_from_path(registry_path).copy()
        registry_for_assembly = registry_from_df(
            reg_df,
            site_id=site_id,
            config_signature=feature_signature,
        )
    else:
        registry = load_or_init_stack_registry(
            site_id,
            artifact_store=feature_pipe.artifact_store,
            config_signature=feature_signature,
        )
        reg_df = stack_registry_frame(registry).copy()
        registry_for_assembly = registry

    if reg_df.empty:
        raise ValueError(
            f"FE stack registry is empty for site={site_id}, "
            f"feature_signature={feature_signature}, registry_path={registry_path}"
        )

    if "base_layer_name" not in reg_df.columns and "layer_name" in reg_df.columns:
        reg_df["base_layer_name"] = reg_df["layer_name"].astype(str).str.split("::").str[0]

    selected_layers = choose_pixel_layer_names_from_df(reg_df, feature_profile)
    if not selected_layers:
        available_sources = sorted(reg_df["source_name"].dropna().astype(str).unique().tolist()) if "source_name" in reg_df.columns else []
        available_families = sorted(reg_df["family_name"].dropna().astype(str).unique().tolist()) if "family_name" in reg_df.columns else []
        sample_layers = sorted(reg_df["base_layer_name"].dropna().astype(str).unique().tolist())[:30]
        raise ValueError(
            "No FE layers selected "
            f"for site={site_id}, profile={feature_profile}. "
            f"available_sources={available_sources}, "
            f"available_families={available_families}, "
            f"sample_base_layers={sample_layers}"
        )

    assembled_cache_path = assembled_bundle_cache_path(
        site_id, feature_signature, feature_profile, resolution_m
    )
    assembled_layers = None if force_refresh else load_cached_assembled_bundle(assembled_cache_path)

    if assembled_layers is None:
        assembled_layers = assemble_selected_layers(
            registry_for_assembly,
            chunk_manifest,
            canonical_grid,
            artifact_store=feature_pipe.artifact_store,
            base_layer_names=selected_layers,
        )
        save_cached_assembled_bundle(assembled_cache_path, assembled_layers)
        print(f"[assembled bundle cache] wrote {assembled_cache_path}")
    else:
        print(f"[assembled bundle cache] hit {assembled_cache_path}")

    label_bundle = read_label_bundle_for_site(
        site_id,
        resolution_m=resolution_m,
        labeling_pipe=labeling_pipe,
    )

    bundle = {
        "site_id": site_id,
        "feature_signature": feature_signature,
        "feature_profile": feature_profile,
        "registry_path": str(registry_path) if registry_path is not None else None,
        "selected_layers": selected_layers,
        "assembled_layers": assembled_layers,
        "cropped_layers": assembled_layers,
        "label_bundle": label_bundle,
        "canonical_grid": canonical_grid,
        "chunk_manifest": chunk_manifest,
        "site_assets": site_assets,
        "feature_pipe": feature_pipe,
        "labeling_pipe": labeling_pipe,
    }

    _SITE_PATCH_CACHE[cache_key] = bundle
    return bundle

def preview_site_ingestion(
    site_id: str,
    *,
    feature_signature: str,
    feature_profile: str = "all_sources",
    registry_path: str | Path | None = None,
) -> pd.DataFrame:
    bundle = materialize_site_feature_label_patch(
        site_id,
        feature_signature=feature_signature,
        feature_profile=feature_profile,
        registry_path=registry_path,
    )

    layer_rows = []
    for layer_name, arr in bundle["cropped_layers"].items():
        layer_rows.append({
            "layer_name": layer_name,
            "shape": tuple(arr.shape),
            "dtype": str(arr.dtype),
            "min": float(np.nanmin(arr)),
            "max": float(np.nanmax(arr)),
            "mean": float(np.nanmean(arr)),
        })

    return pd.DataFrame(layer_rows).sort_values("layer_name").reset_index(drop=True)
```

```python
# Notebook-wired ModelingPipeline ops:
# first smoke-test path = real FE ingestion + real label rasters + cross-site pixel logistic-regression baseline

def build_notebook_modeling_ops():
    def _collect_site_bundles(cfg, modeling_config):
        all_sites = list(cfg.sites)
        site_bundles = {}
        for site_id in all_sites:
            site_bundles[site_id] = materialize_site_feature_label_patch(
                site_id,
                feature_signature=modeling_config.features_config_signature or REAL_FEATURE_SIG,
                feature_profile=modeling_config.feature_profile,
            )
        return site_bundles

    def _flatten_pixels(bundle):
        X = bundle["feature_tensor"].transpose(1, 2, 0).reshape(-1, bundle["feature_tensor"].shape[0])
        y = bundle["label_mask"].reshape(-1).astype(np.uint8)
        w = bundle["confidence"].reshape(-1).astype(np.float32)
        return X, y, w

    def build_site_dataset(*, site_id, artifact_store, modeling_config, cfg):
        site_bundles = _collect_site_bundles(cfg, modeling_config)
        holdout = site_bundles[site_id]
        train_sites = [s for s in cfg.sites if s != site_id and s in site_bundles]

        X_holdout, y_holdout, w_holdout = _flatten_pixels(holdout)

        payload = {
            "site_id": site_id,
            "holdout_site": site_id,
            "train_sites": list(train_sites),
            "n_train_sites": int(len(train_sites)),
            "feature_signature": modeling_config.features_config_signature or REAL_FEATURE_SIG,
            "feature_profile": modeling_config.feature_profile,
            "layer_names": list(holdout["layer_names"]),
            "holdout_feature_shape": list(holdout["feature_tensor"].shape),
            "holdout_label_shape": list(holdout["label_mask"].shape),
            "holdout_positive_fraction": float(y_holdout.mean()),
            "holdout_confidence_mean": float(w_holdout.mean()),
            "label_plot_id": holdout["artifact_row"].get("plot_id"),
            "label_source_version": holdout["artifact_row"].get("source_version"),
        }
        return payload

    def train_model(*, site_id, dataset_bundle, modeling_config, cfg):
        if modeling_config.model_family != "pixel_logreg":
            raise NotImplementedError(
                "Notebook smoke-test wiring currently supports model_family='pixel_logreg'. "
                "Once FE ingestion looks good, you can swap train_model/run_inference to CNN/UNet backends."
            )

        site_bundles = _collect_site_bundles(cfg, modeling_config)
        X_parts, y_parts, w_parts = [], [], []

        for train_site in cfg.sites:
            if train_site == site_id:
                continue
            X_i, y_i, w_i = _flatten_pixels(site_bundles[train_site])
            X_parts.append(X_i)
            y_parts.append(y_i)
            w_parts.append(w_i)

        X_train = np.concatenate(X_parts, axis=0)
        y_train = np.concatenate(y_parts, axis=0)
        w_train = np.concatenate(w_parts, axis=0)

        finite_mask = np.isfinite(X_train).all(axis=1)
        X_train = X_train[finite_mask]
        y_train = y_train[finite_mask]
        w_train = w_train[finite_mask]

        if len(X_train) > MAX_TRAIN_PIXELS:
            rng = np.random.default_rng(RNG_SEED)
            keep_idx = rng.choice(len(X_train), size=MAX_TRAIN_PIXELS, replace=False)
            X_train = X_train[keep_idx]
            y_train = y_train[keep_idx]
            w_train = w_train[keep_idx]

        if np.unique(y_train).size < 2:
            raise ValueError(f"Training labels for holdout={site_id} contain fewer than 2 classes.")

        model = LogisticRegression(
            max_iter=250,
            class_weight="balanced",
            solver="lbfgs",
        )
        sample_weight = w_train if modeling_config.use_confidence_weights else None
        model.fit(X_train, y_train, sample_weight=sample_weight)

        train_probs = model.predict_proba(X_train)[:, 1]
        train_preds = (train_probs >= 0.5).astype(np.uint8)

        train_metrics = {
            "accuracy": float(accuracy_score(y_train, train_preds)),
            "f1": float(f1_score(y_train, train_preds, zero_division=0)),
            "precision": float(precision_score(y_train, train_preds, zero_division=0)),
            "recall": float(recall_score(y_train, train_preds, zero_division=0)),
            "iou": float(jaccard_score(y_train, train_preds, zero_division=0)),
        }

        return {
            "model_family": modeling_config.model_family,
            "feature_profile": modeling_config.feature_profile,
            "layer_names": list(dataset_bundle["layer_names"]),
            "coef": model.coef_.astype(float).tolist(),
            "intercept": model.intercept_.astype(float).tolist(),
            "classes": model.classes_.astype(int).tolist(),
            "train_metrics": train_metrics,
            "n_train_pixels": int(len(X_train)),
            "used_confidence_weights": bool(modeling_config.use_confidence_weights),
        }

    def run_inference(*, site_id, dataset_bundle, model_bundle, modeling_config, cfg):
        holdout = materialize_site_feature_label_patch(
            site_id,
            feature_signature=modeling_config.features_config_signature or REAL_FEATURE_SIG,
            feature_profile=modeling_config.feature_profile,
        )

        X = holdout["feature_tensor"].transpose(1, 2, 0).reshape(-1, holdout["feature_tensor"].shape[0])
        X = np.nan_to_num(X, nan=0.0, posinf=0.0, neginf=0.0)

        coef = np.asarray(model_bundle["coef"], dtype=np.float32)
        intercept = np.asarray(model_bundle["intercept"], dtype=np.float32)

        logits = X @ coef.T
        logits = logits[:, 0] + intercept[0]
        probs = 1.0 / (1.0 + np.exp(-logits))
        prob_map = probs.reshape(holdout["label_mask"].shape).astype(np.float32)

        return {
            "site_id": site_id,
            "probability": prob_map,
            "profile": holdout["profile"],
            "label_mask": holdout["label_mask"].astype(np.uint8),
            "confidence": holdout["confidence"].astype(np.float32),
            "layer_names": holdout["layer_names"],
            "feature_tensor_shape": list(holdout["feature_tensor"].shape),
            "threshold_hint": float(modeling_config.fixed_threshold),
        }

    def calibrate_predictions(*, site_id, dataset_bundle, model_bundle, prediction_bundle, modeling_config, cfg):
        y_true = prediction_bundle["label_mask"].astype(np.uint8).reshape(-1)
        probs = prediction_bundle["probability"].reshape(-1).astype(np.float32)

        if modeling_config.threshold_mode == "fixed":
            thresholds = [float(modeling_config.fixed_threshold)]
        else:
            thresholds = np.linspace(0.1, 0.9, 17)

        rows = []
        for thr in thresholds:
            preds = (probs >= thr).astype(np.uint8)
            rows.append({
                "threshold": float(thr),
                "dice": float((2 * (preds * y_true).sum()) / (preds.sum() + y_true.sum() + 1e-8)),
                "iou": float(jaccard_score(y_true, preds, zero_division=0)),
                "precision": float(precision_score(y_true, preds, zero_division=0)),
                "recall": float(recall_score(y_true, preds, zero_division=0)),
            })

        metrics_df = pd.DataFrame(rows).sort_values(["dice", "iou", "precision"], ascending=False).reset_index(drop=True)
        best = metrics_df.iloc[0].to_dict()

        return {
            "site_id": site_id,
            "threshold": float(best["threshold"]),
            "metrics": {
                "site_id": site_id,
                "best_threshold": float(best["threshold"]),
                "dice": float(best["dice"]),
                "iou": float(best["iou"]),
                "precision": float(best["precision"]),
                "recall": float(best["recall"]),
            },
            "threshold_scan": metrics_df.to_dict(orient="records"),
        }

    def export_prediction_bundle(
        *,
        site_id,
        prediction_bundle,
        artifact_store,
        modeling_config,
        cfg,
        config_signature,
        render_rel_path,
        local_artifact_path,
        push_if_needed,
        prune_if_allowed,
    ):
        profile = prediction_bundle["profile"].copy()
        profile.update(count=1, compress="lzw")

        probability = np.asarray(prediction_bundle["probability"], dtype=np.float32)
        binary = (probability >= float(prediction_bundle.get("threshold_hint", modeling_config.fixed_threshold))).astype(np.uint8)

        def _write(key, arr, dtype):
            rel_path = render_rel_path(key, config_signature=config_signature, site=site_id)
            local_path = local_artifact_path(rel_path)
            local_path.parent.mkdir(parents=True, exist_ok=True)

            write_profile = profile.copy()
            write_profile.update(dtype=dtype, height=arr.shape[0], width=arr.shape[1])

            with rasterio.open(local_path, "w", **write_profile) as dst:
                dst.write(arr.astype(dtype), 1)

            remote_ref = push_if_needed(local_path, key, rel_path)
            prune_if_allowed(local_path, key, rel_path)
            return str(local_path), rel_path, remote_ref

        prob_local, prob_rel, prob_ref = _write("probability_raster", probability, "float32")

        binary_local = binary_rel = binary_ref = None
        if modeling_config.export_binary_predictions:
            binary_local, binary_rel, binary_ref = _write("binary_raster", binary, "uint8")

        return {
            "probability_local_path": prob_local,
            "probability_rel_path": prob_rel,
            "probability_remote_ref": prob_ref,
            "binary_local_path": binary_local,
            "binary_rel_path": binary_rel,
            "binary_remote_ref": binary_ref,
        }

    return ModelingPipelineOps(
        build_site_dataset=build_site_dataset,
        train_model=train_model,
        run_inference=run_inference,
        calibrate_predictions=calibrate_predictions,
        export_prediction_bundle=export_prediction_bundle,
    )

NOTEBOOK_MODELING_OPS = build_notebook_modeling_ops()
```

```python
# Notebook-wired PostprocessingPipeline ops + EvaluationPipeline ops

def build_notebook_postprocessing_ops():
    def _model_prediction_paths(site_id, modeling_sig):
        return {
            "probability_rel": f"modeling/{modeling_sig}/predictions/{site_id}/probability.tif",
            "binary_rel": f"modeling/{modeling_sig}/predictions/{site_id}/binary.tif",
            "calibration_rel": f"modeling/{modeling_sig}/calibration/{site_id}.json",
        }

    def _read_json_if_exists(path: Path):
        if path.exists():
            return json.loads(path.read_text(encoding="utf-8"))
        return None

    def load_site_prediction_bundle(*, site_id, artifact_store, postprocessing_config, cfg):
        paths = _model_prediction_paths(site_id, postprocessing_config.modeling_config_signature)
        prob_path = Path(artifact_store.pull(paths["probability_rel"]))
        with rasterio.open(prob_path) as src:
            probability = src.read(1).astype(np.float32)
            profile = src.profile.copy()

        calibration = {}
        try:
            calib_path = Path(artifact_store.pull(paths["calibration_rel"]))
            calibration = json.loads(calib_path.read_text(encoding="utf-8"))
        except Exception:
            calibration = {}

        return {
            "site_id": site_id,
            "probability": probability,
            "profile": profile,
            "threshold": float(calibration.get("threshold", postprocessing_config.fixed_threshold)),
            "calibration": calibration,
        }

    def load_site_feature_bundle(*, site_id, artifact_store, postprocessing_config, cfg):
        patch = materialize_site_feature_label_patch(
            site_id,
            feature_signature=postprocessing_config.features_config_signature or REAL_FEATURE_SIG,
            feature_profile="all_sources",
        )
        return {
            "site_id": site_id,
            "layer_names": patch["layer_names"],
            "cropped_layers": patch["cropped_layers"],
            "profile": patch["profile"],
        }

    def enrich_object_table(*, site_id, objects_df, feature_bundle, probability, binary_mask, labels, postprocessing_config, cfg):
        if objects_df.empty or feature_bundle is None:
            return objects_df

        layer_names = list(feature_bundle["layer_names"])[:12]
        layers = feature_bundle["cropped_layers"]

        rows = []
        for _, obj in objects_df.iterrows():
            r0 = max(0, int(obj["bbox_min_row"]))
            r1 = min(probability.shape[0], int(obj["bbox_max_row"]))
            c0 = max(0, int(obj["bbox_min_col"]))
            c1 = min(probability.shape[1], int(obj["bbox_max_col"]))

            base = obj.to_dict()
            for layer_name in layer_names:
                patch = np.asarray(layers[layer_name][r0:r1, c0:c1], dtype=np.float32)
                if patch.size == 0:
                    base[f"fe__{layer_name}__mean"] = np.nan
                    base[f"fe__{layer_name}__std"] = np.nan
                else:
                    base[f"fe__{layer_name}__mean"] = float(np.nanmean(patch))
                    base[f"fe__{layer_name}__std"] = float(np.nanstd(patch))
            rows.append(base)

        return pd.DataFrame(rows)

    return PostprocessingPipelineOps(
        load_site_prediction_bundle=load_site_prediction_bundle,
        load_site_feature_bundle=load_site_feature_bundle,
        enrich_object_table=enrich_object_table,
        filter_object_table=None,
    )

NOTEBOOK_POSTPROCESSING_OPS = build_notebook_postprocessing_ops()

def build_notebook_evaluation_ops():
    def _postprocessing_paths(site_id, post_sig):
        return {
            "binary_rel": f"postprocessing/{post_sig}/{site_id}/binary.tif",
            "site_summary_rel": f"postprocessing/{post_sig}/{site_id}/site_summary.json",
            "predicted_objects_rel": f"postprocessing/{post_sig}/{site_id}/predicted_objects.csv",
        }

    def load_label_bundle(*, site_id, artifact_store, evaluation_config, cfg):
        patch = materialize_site_feature_label_patch(
            site_id,
            feature_signature=REAL_FEATURE_SIG,
            feature_profile="all_sources",
        )
        return {
            "site_id": site_id,
            "mask": patch["label_mask"].astype(np.uint8),
            "profile": patch["profile"],
            "objects_df": patch["label_objects_df"],
        }

    def load_prediction_bundle(*, site_id, artifact_store, evaluation_config, cfg):
        paths = _postprocessing_paths(site_id, evaluation_config.postprocessing_config_signature)
        binary_path = Path(artifact_store.pull(paths["binary_rel"]))
        with rasterio.open(binary_path) as src:
            mask = (src.read(1) > 0).astype(np.uint8)
            profile = src.profile.copy()
        return {
            "site_id": site_id,
            "mask": mask,
            "profile": profile,
        }

    def load_site_summary_bundle(*, site_id, artifact_store, evaluation_config, cfg):
        label_patch = materialize_site_feature_label_patch(
            site_id,
            feature_signature=REAL_FEATURE_SIG,
            feature_profile="all_sources",
        )
        label_mask = label_patch["label_mask"].astype(np.uint8)
        label_labels = measure.label(label_mask > 0, connectivity=1)
        label_regions = measure.regionprops(label_labels)

        label_summary = {
            "cover_fraction": float(label_mask.mean()),
            "count_density_per_10k_px": float(len(label_regions) / max(label_mask.size, 1) * 10000.0),
            "mean_object_area_px": float(np.mean([r.area for r in label_regions])) if label_regions else 0.0,
            "median_object_area_px": float(np.median([r.area for r in label_regions])) if label_regions else 0.0,
        }

        paths = _postprocessing_paths(site_id, evaluation_config.postprocessing_config_signature)
        site_summary_path = Path(artifact_store.pull(paths["site_summary_rel"]))
        pred_summary = json.loads(site_summary_path.read_text(encoding="utf-8"))

        return {
            "label_summary": label_summary,
            "pred_summary": pred_summary,
        }

    return EvaluationPipelineOps(
        load_label_bundle=load_label_bundle,
        load_prediction_bundle=load_prediction_bundle,
        load_site_summary_bundle=load_site_summary_bundle,
    )

NOTEBOOK_EVALUATION_OPS = build_notebook_evaluation_ops()
```

```python
# Trial + pipeline bundle builders

def build_downstream_trial_bundle(
    *,
    trial_prefix: str,
    sites: list[str],
    feature_signature: str = REAL_FEATURE_SIG,
    modeling_overrides: dict | None = None,
    postprocessing_overrides: dict | None = None,
    evaluation_overrides: dict | None = None,
):
    local_cfg = make_section_cfg_for_sites(cfg, sites)

    modeling_cfg = ModelingPipelineConfig(
        features_config_signature=feature_signature,
        labeling_config_signature=ACTIVE_LABELING_SIG,
        feature_profile="all_sources",
        model_family="pixel_logreg",
        use_confidence_weights=True,
        threshold_mode="fixed",
        fixed_threshold=0.5,
        export_binary_predictions=True,
    )
    if modeling_overrides:
        for k, v in modeling_overrides.items():
            setattr(modeling_cfg, k, v)

    modeling_pipeline_bundle = ModelingPipeline(
        local_cfg,
        ops=NOTEBOOK_MODELING_OPS,
        pipeline_config=modeling_cfg,
    )

    post_cfg = PostprocessingPipelineConfig(
        modeling_config_signature=modeling_pipeline_bundle.config_signature(),
        features_config_signature=feature_signature,
        threshold_mode="fixed",
        fixed_threshold=modeling_cfg.fixed_threshold,
        cleanup_mode="standard",
        split_mode="watershed",
        enable_object_enrichment=True,
        enable_object_filtering=True,
    )
    if postprocessing_overrides:
        for k, v in postprocessing_overrides.items():
            setattr(post_cfg, k, v)

    post_pipeline_bundle = PostprocessingPipeline(
        local_cfg,
        ops=NOTEBOOK_POSTPROCESSING_OPS,
        pipeline_config=post_cfg,
    )

    eval_cfg = EvaluationPipelineConfig(
        labeling_config_signature=ACTIVE_LABELING_SIG,
        postprocessing_config_signature=post_pipeline_bundle.config_signature(),
    )
    if evaluation_overrides:
        for k, v in evaluation_overrides.items():
            setattr(eval_cfg, k, v)

    evaluation_pipeline_bundle = EvaluationPipeline(
        local_cfg,
        ops=NOTEBOOK_EVALUATION_OPS,
        pipeline_config=eval_cfg,
    )

    trial = grid.get_or_create_trial_for_config(
        pipeline=modeling_pipeline_bundle,
        config_dict=modeling_pipeline_bundle.config_dict(),
        trial_prefix=trial_prefix,
    )

    controller.set_section_config(trial, "modeling", modeling_pipeline_bundle.config_dict())
    controller.set_section_config(trial, "postprocessing", post_pipeline_bundle.config_dict())
    controller.set_section_config(trial, "evaluation", evaluation_pipeline_bundle.config_dict())
    controller.save_trial(trial)

    pipes = {
        "modeling": modeling_pipeline_bundle,
        "postprocessing": post_pipeline_bundle,
        "evaluation": evaluation_pipeline_bundle,
    }
    return trial, pipes, local_cfg

def run_pipeline_bundle_for_trial(trial, pipes):
    global state

    for section_name in ["modeling", "postprocessing", "evaluation"]:
        pipe = pipes[section_name]
        print(f"\n=== RUNNING {section_name.upper()} | trial={trial.trial_id} | config_sig={pipe.config_signature()} ===")
        result, state, runtime_stats = execute_pipeline_section(
            pipe,
            state=state,
            artifact_store=artifact_store,
            push_remote=USE_DRIVE,
        )
        controller.record_section_result(
            trial,
            section_name=section_name,
            config_signature=pipe.config_signature(),
            result=result,
        )
        controller.save_trial(trial)
        controller.save_state(state)

        print({"section": section_name, "status": result.status, "runtime_stats": runtime_stats})

    return controller.load_trial(trial.trial_id)
```

```python
def summarize_stack_registry_file(path: Path) -> dict:
    try:
        payload = json.loads(path.read_text())
    except Exception as e:
        return {
            "path": str(path),
            "exists": path.exists(),
            "valid_json": False,
            "n_layers": None,
            "sources": None,
            "families": None,
            "sample_layer_names": None,
            "columns": None,
            "error": str(e),
        }

    if isinstance(payload, dict):
        layers = payload.get("layers", [])
    elif isinstance(payload, list):
        layers = payload
    else:
        layers = []

    df = pd.DataFrame(layers) if layers else pd.DataFrame()

    out = {
        "path": str(path),
        "exists": path.exists(),
        "valid_json": True,
        "n_layers": len(df),
        "sources": None,
        "families": None,
        "sample_layer_names": None,
        "columns": list(df.columns),
    }

    if not df.empty:
        if "source_name" in df.columns:
            out["sources"] = sorted(df["source_name"].dropna().astype(str).unique().tolist())
        if "family_name" in df.columns:
            out["families"] = sorted(df["family_name"].dropna().astype(str).unique().tolist())
        if "layer_name" in df.columns:
            out["sample_layer_names"] = sorted(df["layer_name"].dropna().astype(str).unique().tolist())[:12]

    return out


def find_candidate_stack_registries(site_id: str):
    candidate_roots = [
        Path(cfg.output.features_root),
        Path(cfg.output.features_root) / "pipeline_runs",
        Path(cfg.output.root) / "features",
        Path("/home/jovyan/work/Dry-shRub/shrub/Final/artifacts/features"),
    ]

    candidates = []
    seen = set()

    for root in candidate_roots:
        print("checking root:", root, "| exists:", root.exists())
        if not root.exists():
            continue

        for p in root.rglob("stack_registry.json"):
            p_str = str(p.resolve())
            if p_str in seen:
                continue
            seen.add(p_str)

            if site_id not in p_str:
                continue

            candidates.append(p)

    rows = [summarize_stack_registry_file(p) for p in sorted(candidates)]

    if not rows:
        return pd.DataFrame(columns=[
            "path", "exists", "valid_json", "n_layers",
            "sources", "families", "sample_layer_names", "columns"
        ])

    return (
        pd.DataFrame(rows)
        .sort_values(["n_layers", "path"], ascending=[False, True], na_position="last")
        .reset_index(drop=True)
    )


SITE = SMOKE_MODEL_SITES[0]
candidate_registry_df = find_candidate_stack_registries(SITE)
display(candidate_registry_df)
```

```python
def load_registry_from_path(path: str | Path) -> pd.DataFrame:
    path = Path(path)
    payload = json.loads(path.read_text())
    layers = payload["layers"] if isinstance(payload, dict) else payload
    df = pd.DataFrame(layers)
    if not df.empty and "layer_name" in df.columns:
        df["base_layer_name"] = df["layer_name"].astype(str).str.split("::").str[0]
    return df
```

```python
def choose_pixel_layer_names_from_df(reg_df: pd.DataFrame, feature_profile: str = "all_sources") -> list[str]:
    if reg_df.empty:
        return []

    profile_to_sources = {
        "naip": {"naip"},
        "naip_als": {"naip", "als"},
        "naip_als_3dep": {"naip", "als", "3dep"},
        "all_sources": {"naip", "als", "3dep", "rap"},
    }
    wanted_sources = profile_to_sources[feature_profile]

    work = reg_df.copy()
    if "base_layer_name" not in work.columns and "layer_name" in work.columns:
        work["base_layer_name"] = work["layer_name"].astype(str).str.split("::").str[0]

    if "source_name" not in work.columns:
        raise ValueError("Registry is missing source_name; cannot select robustly.")

    work = work[work["source_name"].isin(wanted_sources)].copy()
    return sorted(work["base_layer_name"].dropna().astype(str).unique().tolist())
```

```python
def infer_feature_signature_from_registry_path(path: str | Path, site_id: str) -> str | None:
    path = Path(path)
    parts = path.parts

    # Look for: .../stack_registry/{site}__{sig}/stack_registry.json
    for i, part in enumerate(parts):
        if part == "stack_registry" and i + 1 < len(parts):
            next_part = parts[i + 1]
            prefix = f"{site_id}__"
            if next_part.startswith(prefix):
                return next_part[len(prefix):]

    return None

def find_live_feature_pipeline():
    candidates = []

    for name, obj in globals().items():
        try:
            if isinstance(obj, FeaturePipeline):
                candidates.append((name, obj))
        except Exception:
            pass

    print("FeaturePipeline candidates found:")
    for name, obj in candidates:
        print(" -", name, "|", obj)

    if not candidates:
        raise RuntimeError(
            "No live FeaturePipeline object found in globals(). "
            "Run the features pipeline construction cell first, or expose that pipeline object."
        )

    # Prefer obvious names if present
    preferred_names = [
        "feature_pipe",
        "features_pipe",
        "features_pipeline",
        "active_feature_pipeline",
        "feature_pipeline",
    ]
    by_name = dict(candidates)
    for name in preferred_names:
        if name in by_name:
            print("Using preferred FeaturePipeline object:", name)
            return by_name[name]

    # Otherwise just use the first one found
    print("Using first discovered FeaturePipeline object:", candidates[0][0])
    return candidates[0][1]

REAL_FEATURE_SIG = infer_feature_signature_from_registry_path(best_registry_path, SITE)
print("REAL_FEATURE_SIG =", REAL_FEATURE_SIG)
```

```python
from __future__ import annotations

from dataclasses import dataclass, asdict, field
from pathlib import Path
from typing import Any, Iterable
from collections import defaultdict
from copy import deepcopy
import json
import math
import time

import numpy as np
import pandas as pd
from tqdm.auto import tqdm

import rasterio
from rasterio.windows import Window
from rasterio.windows import bounds as window_bounds

from Final.config import default_config
from Final.artifact_store import HybridArtifactStore, LocalArtifactStore, DriveRegistryArtifactStore
from Final.features.pipeline import FeaturePipeline, FeaturePipelineConfig, FeaturePipelineOps
from Final.labeling.pipeline import LabelingPipeline, LabelingPipelineConfig
from Final.features.models import RasterStackRegistry, RasterLayerRecord, ChunkManifest, CanonicalGrid
from Final.features.stack_registry import load_or_init_stack_registry
from Final.features.object_aggregation import stack_registry_frame, load_chunk_layer_array
from Final.features.chunking import build_chunk_manifest, chunk_manifest_frame, chunk_window
from Final.features.canonical_grid import build_canonical_grid_for_site
from Final.features.labeling_bridge import (
    load_labeling_artifacts_summary,
    load_labeling_objects_summary,
    label_objects_for_site,
)

from Final.features.source_naip import prepare_naip_asset, run_naip_chunked_pipeline
from Final.features.source_als import prepare_als_metadata, run_als_chunked_pipeline
from Final.features.source_3dep import prepare_3dep_asset, run_3dep_chunked_pipeline
from Final.features.source_rap import prepare_rap_asset, run_rap_chunked_pipeline
from Final.features.assets import find_cached_rap_asset, find_cached_3dep_asset

@dataclass
class SiteSplitConfig:
    mode: str = "leave_one_site_out"   # "leave_one_site_out", "explicit_site_split"
    train_sites: list[str] = field(default_factory=list)
    val_sites: list[str] = field(default_factory=list)
    test_sites: list[str] = field(default_factory=list)
    held_out_site: str | None = None

@dataclass
class StreamingModelingPlanConfig:
    feature_signature: str
    feature_profile: str = "all_sources"   # "all_sources", "als_3dep", ...
    chunk_sampling_mode: str = "all"       # "all", "subsample"
    max_train_chunks_per_site: int | None = None
    max_val_chunks_per_site: int | None = None
    max_test_chunks_per_site: int | None = None
    use_confidence_weights: bool = True

    # first baselines to support
    baseline_models: tuple[str, ...] = ("sgd_logreg_pixel",)
    # later: ("sgd_logreg_pixel", "small_unet")

@dataclass
class StreamingPostprocessPlanConfig:
    threshold_mode: str = "fixed"          # "fixed", "validation_tuned"
    fixed_threshold: float = 0.5
    cleanup_mode: str = "standard"         # "off", "standard"
    split_mode: str = "connected_components"  # "connected_components", "watershed"
    enable_object_enrichment: bool = False
    enable_object_filtering: bool = False
    enable_count_refinement: bool = False

@dataclass
class StreamingEvaluationPlanConfig:
    composite_weights: dict[str, float] = field(default_factory=lambda: {
        "raster_quality": 0.40,
        "object_quality": 0.30,
        "site_summary": 0.15,
        "generalization": 0.15,
    })
    evaluate_resolutions_m: tuple[float, ...] = (1.0, 2.0, 5.0, 10.0)
    object_iou_threshold: float = 0.3
    centroid_distance_px_threshold: float = 10.0

@dataclass
class StreamingTrialPlan:
    split: SiteSplitConfig
    modeling: StreamingModelingPlanConfig
    postprocessing: StreamingPostprocessPlanConfig
    evaluation: StreamingEvaluationPlanConfig
```

```python
PIPELINE_OBJECT_CACHE: dict[tuple, Any] = {}

def make_downstream_cfg(*, require_feature_hybrid: bool = True, enable_labeling_hybrid: bool = True):
    cfg_local = deepcopy(cfg)

    # features
    cfg_local.features_runtime.storage.enable_local_store = True
    cfg_local.features_runtime.storage.enable_drive_store = True
    cfg_local.features_runtime.storage.use_hybrid_store = True
    cfg_local.features_runtime.storage.fail_if_drive_missing = require_feature_hybrid

    # labeling
    cfg_local.labeling_runtime.storage.enable_local_store = True
    cfg_local.labeling_runtime.storage.enable_drive_store = enable_labeling_hybrid
    cfg_local.labeling_runtime.storage.use_hybrid_store = enable_labeling_hybrid
    cfg_local.labeling_runtime.storage.fail_if_drive_missing = False

    drive_cfg = Path(cfg_local.artifact_store.drive_config_path)
    client_secrets = Path(cfg_local.artifact_store.drive_client_secrets_path)

    print("drive_config_path exists:", drive_cfg, drive_cfg.exists())
    print("client_secrets_path exists:", client_secrets, client_secrets.exists())

    if require_feature_hybrid and (not drive_cfg.exists() or not client_secrets.exists()):
        raise FileNotFoundError(
            "Feature downstream section requires HybridArtifactStore, "
            "but drive config files are missing."
        )

    return cfg_local

def make_feature_ops():
    return FeaturePipelineOps(
        prepare_site_assets=prepare_site_assets,
        build_canonical_grid_for_site=build_canonical_grid_for_site,
        build_chunk_manifest=build_chunk_manifest,
        ensure_source_ready=ensure_source_ready,
        prepare_naip_asset=prepare_naip_asset,
        prepare_als_metadata=prepare_als_metadata,
        prepare_3dep_asset=prepare_3dep_asset,
        prepare_rap_asset=prepare_rap_asset,
        find_cached_rap_asset=find_cached_rap_asset,
        find_cached_3dep_asset=find_cached_3dep_asset,
        run_naip_chunked_pipeline=run_naip_chunked_pipeline,
        run_als_chunked_pipeline=run_als_chunked_pipeline,
        run_3dep_chunked_pipeline=run_3dep_chunked_pipeline,
        run_rap_chunked_pipeline=run_rap_chunked_pipeline,
        label_objects_for_site=label_objects_for_site,
        build_and_persist_object_feature_table=build_and_persist_object_feature_table,
    )

def materialize_feature_pipeline(*, cfg_local, feature_signature=None):
    key = ("features", feature_signature or "active")
    if key in PIPELINE_OBJECT_CACHE:
        return PIPELINE_OBJECT_CACHE[key]

    pipe = FeaturePipeline(
        cfg_local,
        ops=make_feature_ops(),
        pipeline_config=FeaturePipelineConfig(
            canonical_grid_source=cfg_local.features.canonical_grid_source
        ),
    )

    print("materialized feature artifact_store:", type(pipe.artifact_store).__name__)
    if not isinstance(pipe.artifact_store, HybridArtifactStore):
        raise RuntimeError(
            "FeaturePipeline did not materialize with HybridArtifactStore. "
            "Refusing to continue because downstream chunk hydration expects remote-backed FE artifacts."
        )

    PIPELINE_OBJECT_CACHE[key] = pipe
    return pipe

def materialize_labeling_pipeline(*, cfg_local):
    key = ("labeling", "active")
    if key in PIPELINE_OBJECT_CACHE:
        return PIPELINE_OBJECT_CACHE[key]

    pipe = LabelingPipeline(
        cfg_local,
        pipeline_config=LabelingPipelineConfig(),
    )

    print("materialized labeling artifact_store:", type(pipe.artifact_store).__name__)
    PIPELINE_OBJECT_CACHE[key] = pipe
    return pipe

def materialize_downstream_support(*, feature_signature=None, cfg_override=None):
    cfg_local = cfg_override or make_downstream_cfg()
    feature_pipe = materialize_feature_pipeline(
        cfg_local=cfg_local,
        feature_signature=feature_signature,
    )
    labeling_pipe = materialize_labeling_pipeline(
        cfg_local=cfg_local,
    )
    return cfg_local, feature_pipe, labeling_pipe
```

```python
def find_candidate_stack_registries(site_id: str):
    candidate_roots = [
        Path(cfg.output.features_root),
        Path(cfg.output.features_root) / "pipeline_runs",
        Path(cfg.output.root) / "features",
        Path("/home/jovyan/work/Dry-shRub/shrub/Final/artifacts/features"),
    ]

    candidates = []
    seen = set()

    for root in candidate_roots:
        if not root.exists():
            continue
        for p in root.rglob("stack_registry.json"):
            p_str = str(p.resolve())
            if p_str in seen:
                continue
            seen.add(p_str)
            if site_id not in p_str:
                continue
            candidates.append(p)

    rows = [summarize_stack_registry_file(p) for p in sorted(candidates)]
    if not rows:
        return pd.DataFrame(columns=[
            "path", "exists", "valid_json", "n_layers",
            "sources", "families", "sample_layer_names", "columns"
        ])

    return (
        pd.DataFrame(rows)
        .sort_values(["n_layers", "path"], ascending=[False, True], na_position="last")
        .reset_index(drop=True)
    )


def resolve_registry_df_for_site(
    *,
    site_id: str,
    feature_signature: str,
    feature_pipe,
    registry_path: str | Path | None = None,
):
    # 1) explicit path if it still exists
    if registry_path is not None:
        p = Path(registry_path)
        if p.exists():
            payload = json.loads(p.read_text())
            layers = payload["layers"] if isinstance(payload, dict) else payload
            df = pd.DataFrame(layers)
            if not df.empty and "layer_name" in df.columns:
                df["base_layer_name"] = df["layer_name"].astype(str).str.split("::").str[0]
            return df, p

        print(f"[resolve_registry_df_for_site] provided registry_path missing, falling back: {p}")

    # 2) artifact-aware registry load through pipeline store
    registry = load_or_init_stack_registry(
        site_id,
        artifact_store=feature_pipe.artifact_store,
        config_signature=feature_signature,
    )
    reg_df = stack_registry_frame(registry).copy()
    if not reg_df.empty:
        if "layer_name" in reg_df.columns:
            reg_df["base_layer_name"] = reg_df["layer_name"].astype(str).str.split("::").str[0]
        return reg_df, None

    # 3) local candidate search fallback
    candidate_df = find_candidate_stack_registries(site_id)
    if not candidate_df.empty:
        nonempty = candidate_df[candidate_df["n_layers"].fillna(0) > 0].copy()
        if not nonempty.empty:
            best_path = Path(nonempty.iloc[0]["path"])
            payload = json.loads(best_path.read_text())
            layers = payload["layers"] if isinstance(payload, dict) else payload
            df = pd.DataFrame(layers)
            if not df.empty and "layer_name" in df.columns:
                df["base_layer_name"] = df["layer_name"].astype(str).str.split("::").str[0]
            print(f"[resolve_registry_df_for_site] using discovered local registry: {best_path}")
            return df, best_path

    raise FileNotFoundError(
        f"No usable stack registry found for site={site_id}, feature_signature={feature_signature}"
    )

def registry_from_df(reg_df: pd.DataFrame, site_id: str, config_signature: str) -> RasterStackRegistry:
    if reg_df.empty:
        return RasterStackRegistry(site_id=site_id, config_signature=config_signature, layers=[])

    layers = []
    for _, row in reg_df.iterrows():
        payload = row.to_dict()
        allowed = {
            "site_id", "source_name", "family_name", "layer_name",
            "chunk_id", "rel_path", "local_path", "remote_ref",
            "storage_tier", "chunked", "dtype", "shape", "notes",
        }
        payload = {k: v for k, v in payload.items() if k in allowed}
        if "shape" in payload and isinstance(payload["shape"], list):
            payload["shape"] = tuple(payload["shape"])
        layers.append(RasterLayerRecord(**payload))

    return RasterStackRegistry(
        site_id=site_id,
        config_signature=config_signature,
        layers=layers,
    )

def choose_pixel_layer_names_from_df(reg_df: pd.DataFrame, feature_profile: str = "all_sources") -> list[str]:
    if reg_df.empty:
        return []

    profile_to_sources = {
        "naip": {"naip"},
        "naip_als": {"naip", "als"},
        "naip_als_3dep": {"naip", "als", "3dep"},
        "als_3dep": {"als", "3dep"},
        "all_sources": {"naip", "als", "3dep", "rap"},
    }
    if feature_profile not in profile_to_sources:
        raise ValueError(f"Unsupported feature_profile={feature_profile}")

    wanted_sources = profile_to_sources[feature_profile]
    work = reg_df.copy()

    if "base_layer_name" not in work.columns and "layer_name" in work.columns:
        work["base_layer_name"] = work["layer_name"].astype(str).str.split("::").str[0]

    if "source_name" not in work.columns:
        raise ValueError("Registry is missing source_name; cannot select robustly.")

    work = work[work["source_name"].isin(wanted_sources)].copy()
    return sorted(work["base_layer_name"].dropna().astype(str).unique().tolist())

def infer_feature_signature_from_registry_path(path: str | Path, site_id: str) -> str:
    path = Path(path)
    parts = list(path.parts)
    for i in range(len(parts) - 4):
        if parts[i] == "features" and parts[i + 1] == site_id:
            return parts[i + 2]
    raise ValueError(
        f"Could not infer feature signature from path={path}. "
        f"Expected pattern .../features/{site_id}/<config_signature>/stack/stack_registry.json"
    )
```

```python
def resolve_site_split(
    *,
    all_sites: list[str],
    split_cfg: SiteSplitConfig,
) -> dict[str, list[str]]:
    all_sites = list(all_sites)

    if split_cfg.mode == "leave_one_site_out":
        if split_cfg.held_out_site is None:
            raise ValueError("leave_one_site_out requires held_out_site")
        held_out = split_cfg.held_out_site
        if held_out not in all_sites:
            raise ValueError(f"held_out_site={held_out} not in all_sites={all_sites}")

        remaining = [s for s in all_sites if s != held_out]

        # simplest default: last remaining site as validation, rest train
        if split_cfg.val_sites:
            val_sites = list(split_cfg.val_sites)
        else:
            val_sites = remaining[-1:] if remaining else []

        train_sites = [s for s in remaining if s not in val_sites]
        test_sites = [held_out]

    elif split_cfg.mode == "explicit_site_split":
        train_sites = list(split_cfg.train_sites)
        val_sites = list(split_cfg.val_sites)
        test_sites = list(split_cfg.test_sites)

        used = train_sites + val_sites + test_sites
        if len(set(used)) != len(used):
            raise ValueError("explicit_site_split has overlapping site assignments")

        missing = [s for s in used if s not in all_sites]
        if missing:
            raise ValueError(f"Some split sites are not in all_sites: {missing}")

    else:
        raise ValueError(f"Unsupported split mode={split_cfg.mode}")

    return {
        "train_sites": train_sites,
        "val_sites": val_sites,
        "test_sites": test_sites,
    }
```

```python
LABEL_ARTIFACTS_DF = load_labeling_artifacts_summary()
LABEL_OBJECTS_DF = load_labeling_objects_summary()

def normalize_site_id_value(x: Any) -> str:
    return str(x).strip().lower().replace("_", "-").replace(" ", "-")

def label_artifacts_for_site(site_id: str) -> pd.DataFrame:
    if LABEL_ARTIFACTS_DF.empty:
        return pd.DataFrame()
    wanted = normalize_site_id_value(site_id)
    work = LABEL_ARTIFACTS_DF.copy()
    if "_site_id_norm" not in work.columns and "site_id" in work.columns:
        work["_site_id_norm"] = work["site_id"].map(normalize_site_id_value)
    return work[work["_site_id_norm"] == wanted].copy().reset_index(drop=True)

def choose_existing_label_artifact(site_id: str, resolution_m: float = 1.0, labeling_pipe=None) -> pd.Series | None:
    labeling_pipe = labeling_pipe or materialize_downstream_support(feature_signature=REAL_FEATURE_SIG)[2]

    site_df = label_artifacts_for_site(site_id)
    if site_df.empty:
        print(f"[choose_existing_label_artifact] no label artifacts for site={site_id}")
        return None

    work = site_df.copy()

    if "resolution_m" in work.columns:
        work = work[np.isclose(work["resolution_m"].astype(float), float(resolution_m))].copy()

    if work.empty:
        print(f"[choose_existing_label_artifact] no artifacts at resolution={resolution_m} for site={site_id}")
        return None

    def _local_exists(p):
        if p is None or (isinstance(p, float) and np.isnan(p)):
            return False
        try:
            return Path(p).exists()
        except Exception:
            return False

    def _remote_exists(rel_path):
        if rel_path is None or (isinstance(rel_path, float) and np.isnan(rel_path)):
            return False
        try:
            return bool(labeling_pipe.artifact_store.exists(str(rel_path)))
        except Exception:
            return False

    work["binary_exists"] = work["binary_mask_path"].map(_local_exists) if "binary_mask_path" in work.columns else False
    work["confidence_exists"] = work["confidence_mask_path"].map(_local_exists) if "confidence_mask_path" in work.columns else False

    work["binary_remote_exists"] = (
        work["binary_mask_rel_path"].map(_remote_exists)
        if "binary_mask_rel_path" in work.columns else False
    )
    work["confidence_remote_exists"] = (
        work["confidence_mask_rel_path"].map(_remote_exists)
        if "confidence_mask_rel_path" in work.columns else False
    )

    work["source_version_str"] = work["source_version"].astype(str).str.lower() if "source_version" in work.columns else ""
    work["label_variant_str"] = work["label_variant"].astype(str).str.lower() if "label_variant" in work.columns else ""

    work["prefer_revised"] = (
        work["source_version_str"].str.contains("revised", na=False)
        | work["label_variant_str"].str.contains("revised", na=False)
    )

    sort_cols = [
        "prefer_revised",
        "binary_exists",
        "binary_remote_exists",
        "confidence_exists",
        "confidence_remote_exists",
    ]
    ascending = [False, False, False, False, False]

    for c in ["plot_id", "source_version", "label_variant"]:
        if c in work.columns:
            sort_cols.append(c)
            ascending.append(True)

    work = work.sort_values(sort_cols, ascending=ascending).reset_index(drop=True)

    keep_cols = [c for c in [
        "site_id", "plot_id", "source_version", "label_variant", "resolution_m",
        "binary_mask_path", "binary_mask_rel_path",
        "confidence_mask_path", "confidence_mask_rel_path",
        "binary_exists", "binary_remote_exists",
        "confidence_exists", "confidence_remote_exists",
    ] if c in work.columns]
    display(work[keep_cols].head(10))

    # Prefer locally existing binary
    local_ok = work[work["binary_exists"]].copy()
    if not local_ok.empty:
        return local_ok.iloc[0]

    # Then remotely hydratable binary
    remote_ok = work[work["binary_remote_exists"]].copy()
    if not remote_ok.empty:
        return remote_ok.iloc[0]

    return work.iloc[0]

def hydrate_label_artifact_if_needed(
    local_path_str: str | None,
    *,
    rel_path_str: str | None = None,
    labeling_pipe,
):
    # 1) local path already present
    if local_path_str:
        p = Path(local_path_str)
        if p.exists():
            return p
    else:
        p = None

    # 2) explicit rel-path from artifact summary row
    if rel_path_str:
        try:
            if labeling_pipe.artifact_store.exists(str(rel_path_str)):
                # choose a sensible local hydration target
                if p is None:
                    storage_root = (
                        labeling_pipe.artifact_store.local_store.storage_root
                        if hasattr(labeling_pipe.artifact_store, "local_store")
                        else Path(cfg.artifact_store.local_storage_root)
                    )
                    p = storage_root / str(rel_path_str)
                p.parent.mkdir(parents=True, exist_ok=True)
                pulled = labeling_pipe.artifact_store.pull(str(rel_path_str), local_path=p)
                return Path(pulled)
        except Exception as e:
            print(f"[label hydrate] failed for rel_path={rel_path_str}: {e}")

    # 3) fallback: infer rel_path from local path if possible
    if local_path_str:
        p = Path(local_path_str)
        parts = p.parts
        try:
            idx = parts.index("labeling")
            inferred_rel = "/".join(parts[idx:])
            if labeling_pipe.artifact_store.exists(inferred_rel):
                p.parent.mkdir(parents=True, exist_ok=True)
                pulled = labeling_pipe.artifact_store.pull(inferred_rel, local_path=p)
                return Path(pulled)
        except Exception as e:
            print(f"[label hydrate inferred] failed for local_path={local_path_str}: {e}")

    return None

def read_label_bundle_for_site(site_id: str, *, resolution_m: float = 1.0, labeling_pipe=None):
    labeling_pipe = labeling_pipe or materialize_downstream_support(feature_signature=REAL_FEATURE_SIG)[2]

    art = choose_existing_label_artifact(
        site_id,
        resolution_m=resolution_m,
        labeling_pipe=labeling_pipe,
    )
    if art is None:
        raise FileNotFoundError(f"No label artifacts found for site={site_id} at resolution={resolution_m}")

    binary_path = hydrate_label_artifact_if_needed(
        art.get("binary_mask_path"),
        rel_path_str=art.get("binary_mask_rel_path"),
        labeling_pipe=labeling_pipe,
    )
    confidence_path = hydrate_label_artifact_if_needed(
        art.get("confidence_mask_path"),
        rel_path_str=art.get("confidence_mask_rel_path"),
        labeling_pipe=labeling_pipe,
    )

    if binary_path is None or not binary_path.exists():
        raise RuntimeError(
            f"No usable binary label raster found for site={site_id}. "
            f"Chosen local={art.get('binary_mask_path')} | rel={art.get('binary_mask_rel_path')}"
        )

    with rasterio.open(binary_path) as src:
        mask = src.read(1)
        profile = src.profile.copy()
        transform = src.transform
        crs = src.crs

    confidence = None
    if confidence_path is not None and confidence_path.exists():
        with rasterio.open(confidence_path) as src:
            confidence = src.read(1)

    return {
        "site_id": site_id,
        "artifact_row": art,
        "binary_path": binary_path,
        "confidence_path": confidence_path if confidence_path is not None and confidence_path.exists() else None,
        "mask": mask,
        "confidence": confidence,
        "profile": profile,
        "transform": transform,
        "crs": crs,
    }
```

```python
REMOTE_EXISTS_CACHE = {}

def cached_remote_exists(artifact_store, rel_path: str | None) -> bool:
    if not rel_path:
        return False
    key = (type(artifact_store).__name__, str(rel_path))
    if key in REMOTE_EXISTS_CACHE:
        return REMOTE_EXISTS_CACHE[key]
    try:
        ok = bool(artifact_store.exists(str(rel_path)))
    except Exception:
        ok = False
    REMOTE_EXISTS_CACHE[key] = ok
    return ok
```

```python
CHUNK_PLAN_CACHE: dict[tuple, dict[str, Any]] = {}

def build_chunk_availability_df(
    *,
    site_id: str,
    chunk_manifest: ChunkManifest,
    selected_rows: pd.DataFrame,
    artifact_store,
    show_progress: bool = True,
    desc: str | None = None,
) -> pd.DataFrame:
    manifest_df = chunk_manifest_frame(chunk_manifest).copy()
    chunk_ids = manifest_df["chunk_id"].astype(str).tolist()

    work = selected_rows.copy()
    if work.empty:
        return pd.DataFrame([
            {
                "site_id": site_id,
                "chunk_id": chunk_id,
                "n_selected_rows": 0,
                "n_local_rows": 0,
                "n_remote_rows": 0,
                "n_missing_rows": 0,
                "is_chunk_usable": False,
            }
            for chunk_id in chunk_ids
        ])

    if "chunk_id_parsed" not in work.columns:
        work["chunk_id_parsed"] = work["layer_name"].astype(str).str.split("::").str[-1]

    if "local_exists" not in work.columns:
        work["local_exists"] = work["local_path"].map(
            lambda p: bool(p) and Path(p).exists()
        )

    if "remote_exists" not in work.columns:
        unique_rel_paths = sorted(set(
            str(x) for x in work["rel_path"].dropna().astype(str).tolist()
        ))

        rel_map = {}
        rel_iter = tqdm(
            unique_rel_paths,
            desc=(desc or f"{site_id} remote checks"),
            leave=False,
            disable=not show_progress,
        )
        for rel_path in rel_iter:
            rel_map[rel_path] = cached_remote_exists(artifact_store, rel_path)

        work["remote_exists"] = work["rel_path"].map(
            lambda x: rel_map.get(str(x), False) if pd.notna(x) else False
        )

    grouped = (
        work.groupby("chunk_id_parsed")
        .agg(
            n_selected_rows=("layer_name", "count"),
            n_local_rows=("local_exists", "sum"),
            n_remote_rows=("remote_exists", "sum"),
        )
        .reset_index()
        .rename(columns={"chunk_id_parsed": "chunk_id"})
    )

    out = manifest_df[["chunk_id"]].merge(grouped, on="chunk_id", how="left").fillna(0)
    out["site_id"] = site_id
    out["n_selected_rows"] = out["n_selected_rows"].astype(int)
    out["n_local_rows"] = out["n_local_rows"].astype(int)
    out["n_remote_rows"] = out["n_remote_rows"].astype(int)
    out["n_missing_rows"] = out["n_selected_rows"] - out["n_local_rows"] - out["n_remote_rows"]
    out["n_missing_rows"] = out["n_missing_rows"].clip(lower=0).astype(int)
    out["is_chunk_usable"] = (out["n_selected_rows"] > 0) & (out["n_missing_rows"] == 0)

    return out.sort_values("chunk_id").reset_index(drop=True)

def build_site_chunk_plan(
    *,
    site_id: str,
    feature_signature: str,
    feature_profile: str,
    registry_path: str | Path | None = None,
    resolution_m: float = 1.0,
    cfg_override=None,
    force_refresh: bool = False,
    show_progress: bool = True,
) -> dict[str, Any]:
    cache_key = (site_id, feature_signature, feature_profile, str(registry_path), resolution_m)
    if (not force_refresh) and cache_key in CHUNK_PLAN_CACHE:
        return CHUNK_PLAN_CACHE[cache_key]

    t0 = time.perf_counter()
    cfg_local, feature_pipe, labeling_pipe = materialize_downstream_support(
        feature_signature=feature_signature,
        cfg_override=cfg_override,
    )

    t_assets0 = time.perf_counter()
    site_assets = feature_pipe.prepare_site_assets_for_pipeline(site_id, force_refresh=False)
    t_assets1 = time.perf_counter()

    t_grid0 = time.perf_counter()
    canonical_grid = feature_pipe.ops.build_canonical_grid_for_site(
        site_id,
        artifact_store=feature_pipe.artifact_store,
        assets=site_assets,
        force_refresh=False,
    )
    chunk_manifest = feature_pipe.ops.build_chunk_manifest(
        canonical_grid,
        artifact_store=feature_pipe.artifact_store,
        force_refresh=False,
    )
    t_grid1 = time.perf_counter()

    t_reg0 = time.perf_counter()
    reg_df, resolved_registry_path = resolve_registry_df_for_site(
        site_id=site_id,
        feature_signature=feature_signature,
        feature_pipe=feature_pipe,
        registry_path=registry_path,
    )
    registry = registry_from_df(
        reg_df,
        site_id=site_id,
        config_signature=feature_signature,
    )
    t_reg1 = time.perf_counter()

    if reg_df.empty:
        raise ValueError(
            f"Empty stack registry for site={site_id}, feature_signature={feature_signature}, registry_path={registry_path}"
        )

    if "base_layer_name" not in reg_df.columns and "layer_name" in reg_df.columns:
        reg_df["base_layer_name"] = reg_df["layer_name"].astype(str).str.split("::").str[0]

    selected_layers = choose_pixel_layer_names_from_df(reg_df, feature_profile)
    if not selected_layers:
        raise ValueError(f"No layers selected for site={site_id}, feature_profile={feature_profile}")

    selected_rows = reg_df[reg_df["base_layer_name"].isin(selected_layers)].copy()

    t_chunk0 = time.perf_counter()
    chunk_availability_df = build_chunk_availability_df(
        site_id=site_id,
        chunk_manifest=chunk_manifest,
        selected_rows=selected_rows,
        artifact_store=feature_pipe.artifact_store,
        show_progress=show_progress,
        desc=f"{site_id} chunk availability",
    )
    t_chunk1 = time.perf_counter()

    t_label0 = time.perf_counter()
    label_bundle = read_label_bundle_for_site(
        site_id,
        resolution_m=resolution_m,
        labeling_pipe=labeling_pipe,
    )
    t_label1 = time.perf_counter()

    label_shape = tuple(label_bundle["mask"].shape)
    grid_shape = (canonical_grid.height, canonical_grid.width)

    timings = {
        "site_assets_sec": t_assets1 - t_assets0,
        "grid_and_manifest_sec": t_grid1 - t_grid0,
        "registry_sec": t_reg1 - t_reg0,
        "chunk_availability_sec": t_chunk1 - t_chunk0,
        "label_bundle_sec": t_label1 - t_label0,
        "total_sec": time.perf_counter() - t0,
    }

    plan = {
        "site_id": site_id,
        "feature_signature": feature_signature,
        "feature_profile": feature_profile,
        "registry_path": str(registry_path) if registry_path is not None else None,
        "resolved_registry_path": str(resolved_registry_path) if resolved_registry_path is not None else None,
        "feature_pipe": feature_pipe,
        "labeling_pipe": labeling_pipe,
        "site_assets": site_assets,
        "canonical_grid": canonical_grid,
        "chunk_manifest": chunk_manifest,
        "chunk_manifest_df": chunk_manifest_frame(chunk_manifest),
        "registry": registry,
        "registry_df": reg_df,
        "selected_layers": selected_layers,
        "selected_rows": selected_rows,
        "chunk_availability_df": chunk_availability_df,
        "label_bundle": label_bundle,
        "label_shape": label_shape,
        "grid_shape": grid_shape,
        "timings": timings,
    }

    CHUNK_PLAN_CACHE[cache_key] = plan
    return plan
```

```python
def build_streaming_trial_preflight(
    *,
    trial_plan: StreamingTrialPlan,
    all_sites: list[str],
    registry_path_by_site: dict[str, str | Path | None] | None = None,
    cfg_override=None,
    show_progress: bool = True,
) -> dict[str, Any]:
    split = resolve_site_split(all_sites=all_sites, split_cfg=trial_plan.split)

    role_rows = []
    per_site_plans = {}

    ordered_sites = split["train_sites"] + split["val_sites"] + split["test_sites"]
    seen = set()
    ordered_sites = [s for s in ordered_sites if not (s in seen or seen.add(s))]

    outer = tqdm(ordered_sites, desc="Phase 1 preflight", disable=not show_progress)

    site_runtime_rows = []

    for site_id in outer:
        role = (
            "train" if site_id in split["train_sites"]
            else "val" if site_id in split["val_sites"]
            else "test"
        )

        registry_path = None
        if registry_path_by_site is not None:
            registry_path = registry_path_by_site.get(site_id)

        outer.set_postfix({"site": site_id, "role": role})

        plan = build_site_chunk_plan(
            site_id=site_id,
            feature_signature=trial_plan.modeling.feature_signature,
            feature_profile=trial_plan.modeling.feature_profile,
            registry_path=registry_path,
            cfg_override=cfg_override,
            show_progress=show_progress,
        )
        per_site_plans[site_id] = plan

        chunk_df = plan["chunk_availability_df"]
        usable_df = chunk_df[chunk_df["is_chunk_usable"]].copy()

        role_rows.append({
            "site_id": site_id,
            "role": role,
            "feature_signature": trial_plan.modeling.feature_signature,
            "feature_profile": trial_plan.modeling.feature_profile,
            "n_selected_layers": int(len(plan["selected_layers"])),
            "n_total_chunks": int(len(chunk_df)),
            "n_usable_chunks": int(len(usable_df)),
            "n_missing_chunks": int((~chunk_df["is_chunk_usable"]).sum()),
            "grid_height": int(plan["canonical_grid"].height),
            "grid_width": int(plan["canonical_grid"].width),
            "label_shape": str(plan["label_shape"]),
            "grid_shape": str(plan["grid_shape"]),
            "label_binary_path": str(plan["label_bundle"]["binary_path"]),
            "label_confidence_path": str(plan["label_bundle"]["confidence_path"]),
        })

        site_runtime_rows.append({
            "site_id": site_id,
            "role": role,
            **plan["timings"],
        })

    summary = {
        "split": split,
        "role_df": pd.DataFrame(role_rows),
        "site_runtime_df": pd.DataFrame(site_runtime_rows),
        "per_site_plans": per_site_plans,
        "modeling_config": asdict(trial_plan.modeling),
        "postprocessing_config": asdict(trial_plan.postprocessing),
        "evaluation_config": asdict(trial_plan.evaluation),
    }
    return summary
```

```python
def build_chunk_work_plan(preflight_summary: dict[str, Any]) -> pd.DataFrame:
    split = preflight_summary["split"]
    rows = []

    for role_key, role_name in [("train_sites", "train"), ("val_sites", "val"), ("test_sites", "test")]:
        for site_id in split[role_key]:
            plan = preflight_summary["per_site_plans"][site_id]
            chunk_df = plan["chunk_availability_df"]
            usable_df = chunk_df[chunk_df["is_chunk_usable"]].copy()

            for _, row in usable_df.iterrows():
                rows.append({
                    "role": role_name,
                    "site_id": site_id,
                    "chunk_id": row["chunk_id"],
                    "n_selected_rows": int(row["n_selected_rows"]),
                    "n_local_rows": int(row["n_local_rows"]),
                    "n_remote_rows": int(row["n_remote_rows"]),
                    "n_missing_rows": int(row["n_missing_rows"]),
                })

    return pd.DataFrame(rows).reset_index(drop=True)

def summarize_chunk_work_plan(work_df: pd.DataFrame) -> pd.DataFrame:
    if work_df.empty:
        return pd.DataFrame()

    return (
        work_df.groupby(["role", "site_id"])
        .agg(
            n_chunks=("chunk_id", "count"),
            mean_rows_per_chunk=("n_selected_rows", "mean"),
            total_remote_rows=("n_remote_rows", "sum"),
            total_local_rows=("n_local_rows", "sum"),
        )
        .reset_index()
        .sort_values(["role", "site_id"])
    )

def estimate_runtime_from_work_plan(work_df: pd.DataFrame, sec_per_chunk_estimate: float = 3.0) -> pd.DataFrame:
    if work_df.empty:
        return pd.DataFrame()

    summary = (
        work_df.groupby(["role", "site_id"])
        .agg(n_chunks=("chunk_id", "count"))
        .reset_index()
        .sort_values(["role", "site_id"])
    )
    summary["est_seconds"] = summary["n_chunks"] * float(sec_per_chunk_estimate)
    summary["est_minutes"] = summary["est_seconds"] / 60.0
    return summary
```

```python
from collections import OrderedDict
from dataclasses import dataclass
from pathlib import Path
import math
import time

import numpy as np
import pandas as pd
from tqdm.auto import tqdm

import rasterio
from rasterio.windows import Window
from rasterio.windows import transform as window_transform
from rasterio.warp import reproject, Resampling

from sklearn.linear_model import SGDClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    jaccard_score,
)

import torch
import torch.nn as nn
import torch.optim as optim

from Final.modeling.models.base_cnn import SimpleCNN
from Final.modeling.models.unet import CustomUNet, LibraryUNet
```

```python
CHUNK_PATCH_CACHE = OrderedDict()
MAX_CHUNK_PATCH_CACHE_ITEMS = 8

def _chunk_patch_cache_get(key):
    if key not in CHUNK_PATCH_CACHE:
        return None
    value = CHUNK_PATCH_CACHE.pop(key)
    CHUNK_PATCH_CACHE[key] = value
    return value

def _chunk_patch_cache_put(key, value):
    if key in CHUNK_PATCH_CACHE:
        CHUNK_PATCH_CACHE.pop(key)
    CHUNK_PATCH_CACHE[key] = value
    while len(CHUNK_PATCH_CACHE) > MAX_CHUNK_PATCH_CACHE_ITEMS:
        CHUNK_PATCH_CACHE.popitem(last=False)

def get_chunk_record_from_plan(plan: dict, chunk_id: str):
    manifest = plan["chunk_manifest"]

    # Try common manifest shapes
    if hasattr(manifest, "records"):
        records = manifest.records
    elif hasattr(manifest, "chunks"):
        records = manifest.chunks
    else:
        records = None

    if records is None:
        raise AttributeError(
            f"Chunk manifest for site={plan['site_id']} does not expose .records or .chunks"
        )

    # records may be dict-like or iterable
    if isinstance(records, dict):
        if str(chunk_id) in records:
            return records[str(chunk_id)]
        if chunk_id in records:
            return records[chunk_id]
        raise KeyError(f"chunk_id={chunk_id} not found in manifest.records keys")
    else:
        for rec in records:
            rec_chunk_id = getattr(rec, "chunk_id", None)
            if str(rec_chunk_id) == str(chunk_id):
                return rec

    raise KeyError(f"chunk_id={chunk_id} not found for site={plan['site_id']}")

def get_chunk_window_from_record(chunk_record):
    # Prefer the library helper instead of guessing dataframe column names
    try:
        return chunk_window(chunk_record)
    except TypeError:
        # Some versions may want the manifest record only; if this still fails,
        # expose the record fields for debugging.
        fields = getattr(chunk_record, "__dict__", None)
        raise RuntimeError(
            f"Could not build window from chunk record. "
            f"Available fields={fields}"
        )

def get_chunk_rows_for_selected_layers(plan: dict, chunk_id: str) -> pd.DataFrame:
    work = plan["selected_rows"].copy()
    if "chunk_id_parsed" not in work.columns:
        work["chunk_id_parsed"] = work["layer_name"].astype(str).str.split("::").str[-1]
    return work[work["chunk_id_parsed"].astype(str) == str(chunk_id)].copy()
```

```python
CHUNK_NPZ_CACHE = OrderedDict()
MAX_CHUNK_NPZ_CACHE_ITEMS = 16

def _chunk_npz_cache_get(key):
    if key not in CHUNK_NPZ_CACHE:
        return None
    value = CHUNK_NPZ_CACHE.pop(key)
    CHUNK_NPZ_CACHE[key] = value
    return value

def _chunk_npz_cache_put(key, value):
    if key in CHUNK_NPZ_CACHE:
        CHUNK_NPZ_CACHE.pop(key)
    CHUNK_NPZ_CACHE[key] = value
    while len(CHUNK_NPZ_CACHE) > MAX_CHUNK_NPZ_CACHE_ITEMS:
        CHUNK_NPZ_CACHE.popitem(last=False)

def load_family_chunk_npz(
    *,
    family_row: pd.Series,
    artifact_store,
):
    rel_path = family_row.get("rel_path")
    local_path = family_row.get("local_path")
    cache_key = str(rel_path) if pd.notna(rel_path) else str(local_path)

    cached = _chunk_npz_cache_get(cache_key)
    if cached is not None:
        return cached

    # Reuse the same hydration logic path as load_chunk_layer_array, but keep the NPZ in memory
    hydrated_path = None
    if local_path and Path(local_path).exists():
        hydrated_path = Path(local_path)
    elif rel_path and artifact_store.exists(str(rel_path)):
        target_local = Path(local_path) if local_path else None
        if target_local is None:
            if hasattr(artifact_store, "local_store"):
                target_local = Path(artifact_store.local_store.storage_root) / str(rel_path)
            else:
                raise FileNotFoundError(f"No local path and no local_store for rel_path={rel_path}")
        target_local.parent.mkdir(parents=True, exist_ok=True)
        pulled = artifact_store.pull(str(rel_path), local_path=target_local)
        hydrated_path = Path(pulled)
    else:
        raise FileNotFoundError(
            f"Chunk NPZ unavailable. local_path={local_path}, rel_path={rel_path}"
        )

    with np.load(hydrated_path, allow_pickle=False) as data:
        payload = {k: np.asarray(data[k]) for k in data.files}

    # Keep arrays in memory so we don't need the local NPZ anymore
    _chunk_npz_cache_put(cache_key, payload)

    # Optional: prune hydrated file now that payload is in memory
    try:
        if hydrated_path.exists() and "_artifact_staging" in str(hydrated_path):
            hydrated_path.unlink()
    except Exception:
        pass

    return payload

```

```python
def get_chunk_hw_from_record(chunk_record):
    for h_name, w_name in [
        ("height", "width"),
        ("n_rows", "n_cols"),
        ("rows", "cols"),
    ]:
        if hasattr(chunk_record, h_name) and hasattr(chunk_record, w_name):
            return int(getattr(chunk_record, h_name)), int(getattr(chunk_record, w_name))

    fields = getattr(chunk_record, "__dict__", None)
    raise RuntimeError(
        f"Could not infer chunk height/width from chunk record. "
        f"Available fields={fields}"
    )

def materialize_chunk_feature_label_patch(
    *,
    plan: dict,
    chunk_id: str,
    include_confidence: bool = True,
    force_refresh: bool = False,
    show_progress: bool = False,
) -> dict:
    site_id = plan["site_id"]
    feature_signature = plan["feature_signature"]
    feature_profile = plan["feature_profile"]

    cache_key = (
        site_id,
        feature_signature,
        feature_profile,
        str(chunk_id),
        bool(include_confidence),
    )

    if not force_refresh:
        cached = _chunk_patch_cache_get(cache_key)
        if cached is not None:
            return cached

    chunk_record = get_chunk_record_from_plan(plan, chunk_id)
    window = get_chunk_window_from_record(chunk_record)
    height, width = get_chunk_hw_from_record(chunk_record)

        # Group rows by family chunk NPZ path
    chunk_rows = get_chunk_rows_for_selected_layers(plan, chunk_id)
    if chunk_rows.empty:
        raise ValueError(f"No selected feature rows for site={site_id}, chunk_id={chunk_id}")

    layer_names = sorted(chunk_rows["base_layer_name"].dropna().astype(str).unique().tolist())
    feature_arrays = {}

    # Group by the actual NPZ backing file
    group_cols = ["rel_path", "local_path", "source_name", "family_name"]
    grouped = chunk_rows.groupby(group_cols, dropna=False, sort=False)

    group_iter = grouped
    if show_progress:
        group_iter = tqdm(
            list(grouped),
            desc=f"{site_id} {chunk_id} family npz",
            leave=False,
        )

    for group_key, sub in group_iter:
        family_row = sub.iloc[0]
        npz_payload = load_family_chunk_npz(
            family_row=family_row,
            artifact_store=plan["feature_pipe"].artifact_store,
        )

        for _, row in sub.iterrows():
            layer_name_full = str(row["layer_name"])
            base_layer_name = str(row["base_layer_name"])

            # prefer exact layer_name key, then base_layer_name fallback
            if layer_name_full in npz_payload:
                arr = np.asarray(npz_payload[layer_name_full], dtype=np.float32)
            elif base_layer_name in npz_payload:
                arr = np.asarray(npz_payload[base_layer_name], dtype=np.float32)
            else:
                available = list(npz_payload.keys())[:20]
                raise KeyError(
                    f"Layer {layer_name_full} / {base_layer_name} not found in NPZ payload. "
                    f"Available keys sample={available}"
                )

            expected_shape = (height, width)
            if arr.shape != expected_shape:
                raise ValueError(
                    f"Chunk array shape mismatch for site={site_id}, chunk_id={chunk_id}, "
                    f"layer={base_layer_name}: expected {expected_shape}, got {arr.shape}"
                )

            feature_arrays[base_layer_name] = arr

    feature_tensor = np.stack([feature_arrays[name] for name in layer_names], axis=0).astype(np.float32)

    # Reproject label raster into chunk-local destination grid
    label_bundle = plan["label_bundle"]
    dst_transform = window_transform(window, plan["canonical_grid"].transform)
    dst_crs = plan["canonical_grid"].crs

    label_mask = np.zeros((height, width), dtype=np.uint8)
    with rasterio.open(label_bundle["binary_path"]) as src:
        reproject(
            source=rasterio.band(src, 1),
            destination=label_mask,
            src_transform=src.transform,
            src_crs=src.crs,
            dst_transform=dst_transform,
            dst_crs=dst_crs,
            src_nodata=src.nodata,
            dst_nodata=0,
            resampling=Resampling.nearest,
        )

    confidence = np.ones((height, width), dtype=np.float32)
    if include_confidence and label_bundle.get("confidence_path") is not None:
        confidence = np.zeros((height, width), dtype=np.float32)
        with rasterio.open(label_bundle["confidence_path"]) as src:
            reproject(
                source=rasterio.band(src, 1),
                destination=confidence,
                src_transform=src.transform,
                src_crs=src.crs,
                dst_transform=dst_transform,
                dst_crs=dst_crs,
                src_nodata=src.nodata,
                dst_nodata=0.0,
                resampling=Resampling.bilinear,
            )

    out = {
        "site_id": site_id,
        "chunk_id": str(chunk_id),
        "window": window,
        "dst_transform": dst_transform,
        "layer_names": layer_names,
        "feature_arrays": feature_arrays,
        "feature_tensor": feature_tensor,
        "label_mask": label_mask.astype(np.uint8),
        "confidence": confidence.astype(np.float32),
        "height": height,
        "width": width,
    }

    _chunk_patch_cache_put(cache_key, out)
    return out
```

```python
def build_phase2_chunk_work_plan(
    preflight_summary: dict,
    *,
    max_train_chunks_per_site: int | None = None,
    max_val_chunks_per_site: int | None = None,
    max_test_chunks_per_site: int | None = None,
    rng_seed: int = 42,
) -> pd.DataFrame:
    work_df = build_chunk_work_plan(preflight_summary).copy()
    if work_df.empty:
        return work_df

    rng = np.random.default_rng(rng_seed)
    kept_parts = []

    limits = {
        "train": max_train_chunks_per_site,
        "val": max_val_chunks_per_site,
        "test": max_test_chunks_per_site,
    }

    for (role, site_id), sub in work_df.groupby(["role", "site_id"], sort=False):
        sub = sub.copy().reset_index(drop=True)
        limit = limits[role]
        if limit is not None and len(sub) > limit:
            keep_idx = rng.choice(len(sub), size=limit, replace=False)
            sub = sub.iloc[np.sort(keep_idx)].copy().reset_index(drop=True)
        kept_parts.append(sub)

    out = pd.concat(kept_parts, axis=0).reset_index(drop=True)
    return out.sort_values(["role", "site_id", "chunk_id"]).reset_index(drop=True)
```

```python
@dataclass
class StreamingModelState:
    model_name: str
    backend_type: str
    model: Any
    scaler: Any = None
    device: str | None = None
    optimizer: Any = None
    criterion: Any = None
    classes_: tuple[int, ...] = (0, 1)

def make_streaming_model_state(
    *,
    model_name: str,
    in_channels: int,
    learning_rate: float = 1e-3,
    device: str | None = None,
):
    model_name = str(model_name)

    if model_name == "sgd_logreg_pixel":
        model = SGDClassifier(
            loss="log_loss",
            penalty="l2",
            alpha=1e-4,
            random_state=42,
        )
        scaler = StandardScaler(with_mean=True, with_std=True)
        return StreamingModelState(
            model_name=model_name,
            backend_type="sklearn_pixel",
            model=model,
            scaler=scaler,
            classes_=(0, 1),
        )

    device = device or ("cuda" if torch.cuda.is_available() else "cpu")

    if model_name == "simple_cnn":
        model = SimpleCNN(in_channels=in_channels, num_classes=2).to(device)
    elif model_name == "custom_unet":
        model = CustomUNet(in_channels=in_channels, out_channels=1).to(device)
    elif model_name == "library_unet":
        model = LibraryUNet(in_channels=in_channels, out_channels=1).to(device)
    else:
        raise ValueError(f"Unsupported model_name={model_name}")

    optimizer = optim.Adam(model.parameters(), lr=learning_rate)
    criterion = nn.BCEWithLogitsLoss(reduction="none")

    return StreamingModelState(
        model_name=model_name,
        backend_type="torch_segmentation",
        model=model,
        device=device,
        optimizer=optimizer,
        criterion=criterion,
        classes_=(0, 1),
    )
```

```python
def flatten_chunk_for_pixel_model(chunk_patch: dict):
    X = chunk_patch["feature_tensor"].transpose(1, 2, 0).reshape(-1, chunk_patch["feature_tensor"].shape[0])
    y = chunk_patch["label_mask"].reshape(-1).astype(np.uint8)
    w = chunk_patch["confidence"].reshape(-1).astype(np.float32)

    finite = np.isfinite(X).all(axis=1)
    X = X[finite]
    y = y[finite]
    w = w[finite]
    return X, y, w

def build_torch_batch_from_chunk(chunk_patch: dict, device: str):
    X = torch.from_numpy(chunk_patch["feature_tensor"][None, ...]).float().to(device)
    y = torch.from_numpy(chunk_patch["label_mask"][None, None, ...]).float().to(device)
    w = torch.from_numpy(chunk_patch["confidence"][None, None, ...]).float().to(device)
    return X, y, w

def iter_role_site_chunks(work_df: pd.DataFrame, role: str):
    sub = work_df[work_df["role"] == role].copy()
    for site_id, site_df in sub.groupby("site_id", sort=False):
        yield site_id, site_df.reset_index(drop=True)
```

```python
def train_streaming_model(
    *,
    preflight_summary: dict,
    model_name: str,
    learning_rate: float = 1e-3,
    torch_epochs: int = 1,
    max_train_chunks_per_site: int | None = None,
    max_val_chunks_per_site: int | None = None,
    rng_seed: int = 42,
    use_confidence_weights: bool = True,
    show_progress: bool = True,
):
    work_df = build_phase2_chunk_work_plan(
        preflight_summary,
        max_train_chunks_per_site=max_train_chunks_per_site,
        max_val_chunks_per_site=max_val_chunks_per_site,
        max_test_chunks_per_site=None,
        rng_seed=rng_seed,
    )

    train_df = work_df[work_df["role"] == "train"].copy().reset_index(drop=True)
    val_df = work_df[work_df["role"] == "val"].copy().reset_index(drop=True)

    if train_df.empty:
        raise ValueError("No train chunks available for Phase 2")

    first_row = train_df.iloc[0]
    first_plan = preflight_summary["per_site_plans"][first_row["site_id"]]
    first_chunk = materialize_chunk_feature_label_patch(
        plan=first_plan,
        chunk_id=first_row["chunk_id"],
        show_progress=False,
    )
    in_channels = int(first_chunk["feature_tensor"].shape[0])

    state = make_streaming_model_state(
        model_name=model_name,
        in_channels=in_channels,
        learning_rate=learning_rate,
    )

    if state.backend_type == "sklearn_pixel":
        # scaler pass
        outer_sites = list(iter_role_site_chunks(work_df, "train"))
        outer_bar = tqdm(outer_sites, desc=f"{model_name} scaler sites", disable=not show_progress)

        for site_id, site_df in outer_bar:
            outer_bar.set_postfix({"site": site_id, "chunks": len(site_df)})
            chunk_bar = tqdm(
                site_df.itertuples(index=False),
                total=len(site_df),
                desc=f"{site_id} scaler chunks",
                leave=False,
                disable=not show_progress,
            )
            for row in chunk_bar:
                plan = preflight_summary["per_site_plans"][row.site_id]
                chunk_patch = materialize_chunk_feature_label_patch(
                    plan=plan,
                    chunk_id=row.chunk_id,
                    show_progress=False,
                )
                X, y, w = flatten_chunk_for_pixel_model(chunk_patch)
                if len(X) == 0:
                    continue
                state.scaler.partial_fit(X)

        # train pass
        outer_bar = tqdm(outer_sites, desc=f"{model_name} train sites", disable=not show_progress)
        fitted_once = False

        for site_id, site_df in outer_bar:
            outer_bar.set_postfix({"site": site_id, "chunks": len(site_df)})
            chunk_bar = tqdm(
                site_df.itertuples(index=False),
                total=len(site_df),
                desc=f"{site_id} train chunks",
                leave=False,
                disable=not show_progress,
            )
            for row in chunk_bar:
                plan = preflight_summary["per_site_plans"][row.site_id]
                chunk_patch = materialize_chunk_feature_label_patch(
                    plan=plan,
                    chunk_id=row.chunk_id,
                    show_progress=False,
                )
                X, y, w = flatten_chunk_for_pixel_model(chunk_patch)
                if len(X) == 0 or np.unique(y).size < 2:
                    continue

                Xs = state.scaler.transform(X)
                sw = w if use_confidence_weights else None

                if not fitted_once:
                    state.model.partial_fit(Xs, y, classes=np.array(state.classes_), sample_weight=sw)
                    fitted_once = True
                else:
                    state.model.partial_fit(Xs, y, sample_weight=sw)

                chunk_bar.set_postfix({
                    "site": row.site_id,
                    "chunk": row.chunk_id,
                    "n": len(Xs),
                })

        return state


def score_streaming_validation(
    *,
    preflight_summary: dict,
    state: StreamingModelState,
    max_val_chunks_per_site: int | None = None,
    rng_seed: int = 42,
    show_progress: bool = True,
):
    work_df = build_phase2_chunk_work_plan(
        preflight_summary,
        max_train_chunks_per_site=None,
        max_val_chunks_per_site=max_val_chunks_per_site,
        max_test_chunks_per_site=None,
        rng_seed=rng_seed,
    )
    val_df = work_df[work_df["role"] == "val"].copy().reset_index(drop=True)
    if val_df.empty:
        return pd.DataFrame()

    rows = []
    outer_sites = list(iter_role_site_chunks(work_df, "val"))
    outer_bar = tqdm(outer_sites, desc=f"{state.model_name} val sites", disable=not show_progress)

    for site_id, site_df in outer_bar:
        outer_bar.set_postfix({"site": site_id, "chunks": len(site_df)})
        chunk_bar = tqdm(
            site_df.itertuples(index=False),
            total=len(site_df),
            desc=f"{site_id} val chunks",
            leave=False,
            disable=not show_progress,
        )

        for row in chunk_bar:
            plan = preflight_summary["per_site_plans"][row.site_id]
            chunk_patch = materialize_chunk_feature_label_patch(
                plan=plan,
                chunk_id=row.chunk_id,
                show_progress=False,
            )

            y_true = chunk_patch["label_mask"].reshape(-1).astype(np.uint8)

            if state.backend_type == "sklearn_pixel":
                X, _, _ = flatten_chunk_for_pixel_model(chunk_patch)
                Xs = state.scaler.transform(X)
                probs = state.model.predict_proba(Xs)[:, 1]
                preds = (probs >= 0.5).astype(np.uint8)
                y_eval = y_true[:len(preds)]
            else:
                state.model.eval()
                with torch.no_grad():
                    X, _, _ = build_torch_batch_from_chunk(chunk_patch, state.device)
                    logits = state.model(X)
                    if logits.ndim == 4 and logits.shape[1] == 2:
                        logits = logits[:, 1:2, :, :]
                    elif logits.ndim == 3:
                        logits = logits[:, None, :, :]
                    probs = torch.sigmoid(logits).cpu().numpy().reshape(-1)
                    preds = (probs >= 0.5).astype(np.uint8)
                    y_eval = y_true

            rows.append({
                "site_id": row.site_id,
                "chunk_id": row.chunk_id,
                "accuracy": float(accuracy_score(y_eval, preds)),
                "f1": float(f1_score(y_eval, preds, zero_division=0)),
                "precision": float(precision_score(y_eval, preds, zero_division=0)),
                "recall": float(recall_score(y_eval, preds, zero_division=0)),
                "iou": float(jaccard_score(y_eval, preds, zero_division=0)),
            })

    return pd.DataFrame(rows)
```

```python
def create_probability_raster_writer(output_path: Path, canonical_grid):
    output_path.parent.mkdir(parents=True, exist_ok=True)
    profile = canonical_grid.profile(dtype="float32", count=1, nodata=np.nan)
    return rasterio.open(output_path, "w", **profile)

def write_test_site_probability_raster(
    *,
    preflight_summary: dict,
    state: StreamingModelState,
    test_site_id: str,
    output_dir: str | Path,
    max_test_chunks_per_site: int | None = None,
    rng_seed: int = 42,
    show_progress: bool = True,
):
    work_df = build_phase2_chunk_work_plan(
        preflight_summary,
        max_train_chunks_per_site=None,
        max_val_chunks_per_site=None,
        max_test_chunks_per_site=max_test_chunks_per_site,
        rng_seed=rng_seed,
    )
    test_df = work_df[(work_df["role"] == "test") & (work_df["site_id"] == test_site_id)].copy().reset_index(drop=True)
    if test_df.empty:
        raise ValueError(f"No test chunks available for site={test_site_id}")

    plan = preflight_summary["per_site_plans"][test_site_id]
    canonical_grid = plan["canonical_grid"]

    output_dir = Path(output_dir)
    prob_path = output_dir / f"{test_site_id}__{state.model_name}__probability.tif"

    with create_probability_raster_writer(prob_path, canonical_grid) as dst:
        bar = tqdm(
            test_df.itertuples(index=False),
            total=len(test_df),
            desc=f"{state.model_name} {test_site_id} test chunks",
            disable=not show_progress,
        )
        for row in bar:
            chunk_patch = materialize_chunk_feature_label_patch(
                plan=plan,
                chunk_id=row.chunk_id,
                show_progress=False,
            )
            window = chunk_patch["window"]

            if state.backend_type == "sklearn_pixel":
                X, _, _ = flatten_chunk_for_pixel_model(chunk_patch)
                Xs = state.scaler.transform(X)
                probs = state.model.predict_proba(Xs)[:, 1]
                prob_map = probs.reshape(chunk_patch["height"], chunk_patch["width"]).astype(np.float32)
            else:
                state.model.eval()
                with torch.no_grad():
                    X, _, _ = build_torch_batch_from_chunk(chunk_patch, state.device)
                    logits = state.model(X)
                    if logits.ndim == 4 and logits.shape[1] == 2:
                        logits = logits[:, 1:2, :, :]
                    elif logits.ndim == 3:
                        logits = logits[:, None, :, :]
                    prob_map = torch.sigmoid(logits).cpu().numpy()[0, 0].astype(np.float32)

            dst.write(prob_map, 1, window=window)
            bar.set_postfix({"chunk": row.chunk_id, "shape": f"{chunk_patch['height']}x{chunk_patch['width']}"})

    return prob_path
```

```python
def run_phase2_streaming_modeling(
    *,
    preflight_summary: dict,
    model_name: str,
    output_dir: str | Path,
    learning_rate: float = 1e-3,
    torch_epochs: int = 1,
    max_train_chunks_per_site: int | None = None,
    max_val_chunks_per_site: int | None = None,
    max_test_chunks_per_site: int | None = None,
    rng_seed: int = 42,
    show_progress: bool = True,
):
    t0 = time.perf_counter()

    print(f"\n=== PHASE 2 | model={model_name} ===")
    print("train sites:", preflight_summary["split"]["train_sites"])
    print("val sites:", preflight_summary["split"]["val_sites"])
    print("test sites:", preflight_summary["split"]["test_sites"])

    t_train0 = time.perf_counter()
    state = train_streaming_model(
        preflight_summary=preflight_summary,
        model_name=model_name,
        learning_rate=learning_rate,
        torch_epochs=torch_epochs,
        max_train_chunks_per_site=max_train_chunks_per_site,
        max_val_chunks_per_site=max_val_chunks_per_site,
        rng_seed=rng_seed,
        use_confidence_weights=preflight_summary["modeling_config"]["use_confidence_weights"],
        show_progress=show_progress,
    )
    t_train1 = time.perf_counter()

    t_val0 = time.perf_counter()
    val_df = score_streaming_validation(
        preflight_summary=preflight_summary,
        state=state,
        max_val_chunks_per_site=max_val_chunks_per_site,
        rng_seed=rng_seed,
        show_progress=show_progress,
    )
    t_val1 = time.perf_counter()

    split = preflight_summary["split"]
    test_sites = list(split["test_sites"])
    written = []

    t_inf0 = time.perf_counter()
    for test_site_id in test_sites:
        prob_path = write_test_site_probability_raster(
            preflight_summary=preflight_summary,
            state=state,
            test_site_id=test_site_id,
            output_dir=output_dir,
            max_test_chunks_per_site=max_test_chunks_per_site,
            rng_seed=rng_seed,
            show_progress=show_progress,
        )
        written.append({
            "site_id": test_site_id,
            "probability_raster_path": str(prob_path),
        })
    t_inf1 = time.perf_counter()

    total_sec = time.perf_counter() - t0

    summary = {
        "model_name": model_name,
        "backend_type": state.backend_type,
        "validation_metrics_df": val_df,
        "written_outputs_df": pd.DataFrame(written),
        "runtime_seconds": total_sec,
        "timings": {
            "train_sec": t_train1 - t_train0,
            "val_sec": t_val1 - t_val0,
            "infer_sec": t_inf1 - t_inf0,
            "total_sec": total_sec,
        },
    }

    print("Phase 2 timings:", summary["timings"])
    return state, summary
```

### Postprocessing

```python
from collections import defaultdict
from pathlib import Path
import json
import time

import numpy as np
import pandas as pd
from tqdm.auto import tqdm

import rasterio
from rasterio.windows import Window

from scipy import ndimage as ndi
from skimage import measure, morphology, segmentation
```

```python
DEFAULT_POSTPROCESS_ENRICHMENT_LAYERS = (
    "als_chm_max",
    "als_height_mean",
    "als_height_std",
    "als_local_relief",
    "terrain_slope",
    "terrain_tpi",
    "rap_shrub_fraction_prior",
)

def get_postprocess_threshold(preflight_summary: dict) -> float:
    cfg = preflight_summary["postprocessing_config"]
    if cfg["threshold_mode"] == "fixed":
        return float(cfg["fixed_threshold"])
    raise NotImplementedError("validation_tuned thresholding will be added after Phase 4 metrics are in place.")

def canonical_pixel_area_m2(plan: dict) -> float:
    transform = plan["canonical_grid"].transform
    return float(abs(transform.a * transform.e))
```

```python
def probability_output_map_from_phase2(phase2_summary: dict) -> dict[str, Path]:
    df = phase2_summary["written_outputs_df"].copy()
    out = {}
    for _, row in df.iterrows():
        out[str(row["site_id"])] = Path(row["probability_raster_path"])
    return out

def open_probability_raster(site_id: str, phase2_summary: dict):
    prob_map = probability_output_map_from_phase2(phase2_summary)
    if site_id not in prob_map:
        raise KeyError(f"No probability raster found for site={site_id}")
    return rasterio.open(prob_map[site_id])

def create_single_band_writer(path: Path, profile: dict, dtype: str, nodata=None):
    path.parent.mkdir(parents=True, exist_ok=True)
    out_profile = profile.copy()
    out_profile.update(count=1, dtype=dtype)
    if nodata is not None:
        out_profile["nodata"] = nodata
    return rasterio.open(path, "w", **out_profile)
```

```python
def threshold_probability_raster_chunkwise(
    *,
    plan: dict,
    probability_raster_path: str | Path,
    threshold: float,
    output_dir: str | Path,
    show_progress: bool = True,
):
    site_id = plan["site_id"]
    probability_raster_path = Path(probability_raster_path)
    output_dir = Path(output_dir)

    raw_binary_path = output_dir / f"{site_id}__raw_binary.tif"

    with rasterio.open(probability_raster_path) as src:
        profile = src.profile.copy()

        with create_single_band_writer(raw_binary_path, profile, dtype="uint8", nodata=0) as dst:
            chunk_df = plan["chunk_availability_df"].copy()
            chunk_df = chunk_df[chunk_df["is_chunk_usable"]].copy().reset_index(drop=True)

            bar = tqdm(
                chunk_df.itertuples(index=False),
                total=len(chunk_df),
                desc=f"{site_id} threshold chunks",
                disable=not show_progress,
            )

            for row in bar:
                chunk_patch = materialize_chunk_feature_label_patch(
                    plan=plan,
                    chunk_id=row.chunk_id,
                    show_progress=False,
                )
                window = chunk_patch["window"]
                prob = src.read(1, window=window).astype(np.float32)
                binary = (prob >= float(threshold)).astype(np.uint8)
                dst.write(binary, 1, window=window)
                bar.set_postfix({"chunk": row.chunk_id, "thr": threshold})

    return raw_binary_path
```

```python
def cleanup_binary_mask(
    binary_mask: np.ndarray,
    *,
    cleanup_mode: str = "standard",
    pixel_area_m2: float = 0.36,
    min_object_area_m2: float = 1.0,
    min_hole_area_m2: float = 0.5,
) -> np.ndarray:
    binary = binary_mask.astype(bool)

    if cleanup_mode == "off":
        return binary.astype(np.uint8)

    if cleanup_mode != "standard":
        raise ValueError(f"Unsupported cleanup_mode={cleanup_mode}")

    min_obj_px = max(1, int(round(min_object_area_m2 / max(pixel_area_m2, 1e-8))))
    min_hole_px = max(1, int(round(min_hole_area_m2 / max(pixel_area_m2, 1e-8))))

    cleaned = morphology.remove_small_objects(binary, min_size=min_obj_px, connectivity=2)
    cleaned = morphology.remove_small_holes(cleaned, area_threshold=min_hole_px, connectivity=2)
    cleaned = morphology.opening(cleaned, morphology.disk(1))
    cleaned = morphology.closing(cleaned, morphology.disk(1))

    return cleaned.astype(np.uint8)

def split_binary_into_objects(
    binary_mask: np.ndarray,
    *,
    split_mode: str = "connected_components",
) -> np.ndarray:
    binary = binary_mask.astype(bool)

    if split_mode == "connected_components":
        labels = measure.label(binary, connectivity=2)
        return labels.astype(np.int32)

    if split_mode == "watershed":
        distance = ndi.distance_transform_edt(binary)
        local_max = morphology.local_maxima(distance)
        markers = measure.label(local_max, connectivity=2)
        labels = segmentation.watershed(-distance, markers=markers, mask=binary)
        return labels.astype(np.int32)

    raise ValueError(f"Unsupported split_mode={split_mode}")
```

```python
def postprocess_site_binary_and_labels(
    *,
    plan: dict,
    probability_raster_path: str | Path,
    post_cfg: dict,
    output_dir: str | Path,
    show_progress: bool = True,
):
    site_id = plan["site_id"]
    output_dir = Path(output_dir)
    threshold = float(post_cfg["fixed_threshold"]) if post_cfg["threshold_mode"] == "fixed" else get_postprocess_threshold({"postprocessing_config": post_cfg})

    raw_binary_path = threshold_probability_raster_chunkwise(
        plan=plan,
        probability_raster_path=probability_raster_path,
        threshold=threshold,
        output_dir=output_dir,
        show_progress=show_progress,
    )

    with rasterio.open(raw_binary_path) as src:
        raw_binary = src.read(1).astype(np.uint8)
        profile = src.profile.copy()

    cleaned_binary = cleanup_binary_mask(
        raw_binary,
        cleanup_mode=post_cfg["cleanup_mode"],
        pixel_area_m2=canonical_pixel_area_m2(plan),
        min_object_area_m2=1.0,
        min_hole_area_m2=0.5,
    )

    labels = split_binary_into_objects(
        cleaned_binary,
        split_mode=post_cfg["split_mode"],
    )

    cleaned_binary_path = output_dir / f"{site_id}__cleaned_binary.tif"
    labels_path = output_dir / f"{site_id}__object_labels.tif"

    with create_single_band_writer(cleaned_binary_path, profile, dtype="uint8", nodata=0) as dst:
        dst.write(cleaned_binary.astype(np.uint8), 1)

    label_profile = profile.copy()
    label_profile.update(dtype="int32", nodata=0)
    with create_single_band_writer(labels_path, label_profile, dtype="int32", nodata=0) as dst:
        dst.write(labels.astype(np.int32), 1)

    return {
        "site_id": site_id,
        "threshold": threshold,
        "raw_binary_path": raw_binary_path,
        "cleaned_binary_path": cleaned_binary_path,
        "labels_path": labels_path,
        "cleaned_binary": cleaned_binary,
        "labels": labels,
        "profile": profile,
    }
```

```python
def build_object_geometry_table(
    *,
    plan: dict,
    labels: np.ndarray,
) -> pd.DataFrame:
    regions = measure.regionprops(labels)
    pixel_area = canonical_pixel_area_m2(plan)
    transform = plan["canonical_grid"].transform

    rows = []
    for r in regions:
        cy, cx = r.centroid
        x, y = rasterio.transform.xy(transform, cy, cx, offset="center")

        perimeter_px = float(r.perimeter) if np.isfinite(r.perimeter) else 0.0
        area_px = int(r.area)
        compactness = float((4.0 * math.pi * area_px) / ((perimeter_px ** 2) + 1e-8))

        rows.append({
            "object_id": int(r.label),
            "area_px": area_px,
            "area_m2": float(area_px * pixel_area),
            "perimeter_px": perimeter_px,
            "centroid_row": float(cy),
            "centroid_col": float(cx),
            "centroid_x": float(x),
            "centroid_y": float(y),
            "bbox_min_row": int(r.bbox[0]),
            "bbox_min_col": int(r.bbox[1]),
            "bbox_max_row": int(r.bbox[2]),
            "bbox_max_col": int(r.bbox[3]),
            "eccentricity": float(getattr(r, "eccentricity", np.nan)),
            "solidity": float(getattr(r, "solidity", np.nan)),
            "extent": float(getattr(r, "extent", np.nan)),
            "equivalent_diameter": float(getattr(r, "equivalent_diameter_area", np.nan)),
            "major_axis_length": float(getattr(r, "axis_major_length", getattr(r, "major_axis_length", np.nan))),
            "minor_axis_length": float(getattr(r, "axis_minor_length", getattr(r, "minor_axis_length", np.nan))),
            "compactness": compactness,
        })

    return pd.DataFrame(rows).sort_values("object_id").reset_index(drop=True)
```

```python
def choose_enrichment_layers(plan: dict, requested_layers=DEFAULT_POSTPROCESS_ENRICHMENT_LAYERS):
    selected = set(plan["selected_layers"])
    return [name for name in requested_layers if name in selected]

def enrich_object_table_streaming(
    *,
    plan: dict,
    object_df: pd.DataFrame,
    labels: np.ndarray,
    probability_raster_path: str | Path,
    enrichment_layers: tuple[str, ...] = DEFAULT_POSTPROCESS_ENRICHMENT_LAYERS,
    show_progress: bool = True,
):
    if object_df.empty:
        return object_df.copy()

    site_id = plan["site_id"]
    enrich_layers = choose_enrichment_layers(plan, enrichment_layers)

    def _make_accumulator():
        d = {
            "pixel_count_seen": 0,
            "prob_sum": 0.0,
            "prob_max": -np.inf,
            "conf_sum": 0.0,
        }
        for layer_name in enrich_layers:
            d[f"{layer_name}__sum"] = 0.0
            d[f"{layer_name}__max"] = -np.inf
        return d

    acc = defaultdict(_make_accumulator)

    chunk_df = plan["chunk_availability_df"].copy()
    chunk_df = chunk_df[chunk_df["is_chunk_usable"]].copy().reset_index(drop=True)

    with rasterio.open(probability_raster_path) as src:
        bar = tqdm(
            chunk_df.itertuples(index=False),
            total=len(chunk_df),
            desc=f"{site_id} object enrichment chunks",
            disable=not show_progress,
        )

        for row in bar:
            chunk_patch = materialize_chunk_feature_label_patch(
                plan=plan,
                chunk_id=row.chunk_id,
                show_progress=False,
            )
            window = chunk_patch["window"]

            r0 = int(window.row_off)
            c0 = int(window.col_off)
            r1 = r0 + int(window.height)
            c1 = c0 + int(window.width)

            local_labels = labels[r0:r1, c0:c1]
            if local_labels.size == 0:
                continue

            prob = src.read(1, window=window).astype(np.float32)
            conf = chunk_patch["confidence"].astype(np.float32)

            obj_ids = np.unique(local_labels)
            obj_ids = obj_ids[obj_ids > 0]

            for obj_id in obj_ids:
                mask = (local_labels == obj_id)
                n = int(mask.sum())
                if n == 0:
                    continue

                d = acc[int(obj_id)]
                d["pixel_count_seen"] += n
                d["prob_sum"] += float(prob[mask].sum())
                d["prob_max"] = max(d["prob_max"], float(prob[mask].max()))
                d["conf_sum"] += float(conf[mask].sum())

                for layer_name in enrich_layers:
                    arr = chunk_patch["feature_arrays"][layer_name]
                    d[f"{layer_name}__sum"] += float(arr[mask].sum())
                    d[f"{layer_name}__max"] = max(d[f"{layer_name}__max"], float(arr[mask].max()))

    out = object_df.copy()

    out["mean_probability"] = out["object_id"].map(
        lambda oid: acc[int(oid)]["prob_sum"] / max(acc[int(oid)]["pixel_count_seen"], 1)
    )
    out["max_probability"] = out["object_id"].map(
        lambda oid: acc[int(oid)]["prob_max"] if np.isfinite(acc[int(oid)]["prob_max"]) else np.nan
    )
    out["mean_confidence"] = out["object_id"].map(
        lambda oid: acc[int(oid)]["conf_sum"] / max(acc[int(oid)]["pixel_count_seen"], 1)
    )

    for layer_name in enrich_layers:
        out[f"{layer_name}__mean"] = out["object_id"].map(
            lambda oid: acc[int(oid)][f"{layer_name}__sum"] / max(acc[int(oid)]["pixel_count_seen"], 1)
        )
        out[f"{layer_name}__max"] = out["object_id"].map(
            lambda oid: acc[int(oid)][f"{layer_name}__max"] if np.isfinite(acc[int(oid)][f"{layer_name}__max"]) else np.nan
        )

    return out
```

```python
def build_site_summary(
    *,
    plan: dict,
    cleaned_binary: np.ndarray,
    object_df: pd.DataFrame,
) -> dict:
    pixel_area = canonical_pixel_area_m2(plan)
    site_area_m2 = float(cleaned_binary.size * pixel_area)
    shrub_area_m2 = float(cleaned_binary.sum() * pixel_area)

    summary = {
        "site_id": plan["site_id"],
        "grid_height": int(plan["canonical_grid"].height),
        "grid_width": int(plan["canonical_grid"].width),
        "pixel_area_m2": pixel_area,
        "site_area_m2": site_area_m2,
        "shrub_area_m2": shrub_area_m2,
        "cover_fraction": float(cleaned_binary.mean()),
        "n_objects": int(len(object_df)),
        "count_density_per_ha": float(len(object_df) / max(site_area_m2 / 10000.0, 1e-8)),
        "mean_object_area_m2": float(object_df["area_m2"].mean()) if not object_df.empty else 0.0,
        "median_object_area_m2": float(object_df["area_m2"].median()) if not object_df.empty else 0.0,
        "mean_object_probability": float(object_df["mean_probability"].mean()) if "mean_probability" in object_df.columns and not object_df.empty else np.nan,
    }

    if "als_chm_max__mean" in object_df.columns:
        summary["mean_object_height_proxy"] = float(object_df["als_chm_max__mean"].mean())

    return summary
```

```python
def run_phase3_postprocessing_for_site(
    *,
    preflight_summary: dict,
    phase2_summary: dict,
    site_id: str,
    output_dir: str | Path,
    enrichment_layers: tuple[str, ...] = DEFAULT_POSTPROCESS_ENRICHMENT_LAYERS,
    show_progress: bool = True,
):
    split = preflight_summary["split"]
    if site_id not in split["test_sites"]:
        print(f"[phase3] site={site_id} is not in test_sites; proceeding anyway.")

    plan = preflight_summary["per_site_plans"][site_id]
    post_cfg = preflight_summary["postprocessing_config"]
    prob_map = probability_output_map_from_phase2(phase2_summary)
    if site_id not in prob_map:
        raise KeyError(f"No Phase 2 probability raster found for site={site_id}")

    site_out = Path(output_dir) / site_id
    site_out.mkdir(parents=True, exist_ok=True)

    t0 = time.perf_counter()

    stage1 = postprocess_site_binary_and_labels(
        plan=plan,
        probability_raster_path=prob_map[site_id],
        post_cfg=post_cfg,
        output_dir=site_out,
        show_progress=show_progress,
    )

    geom_df = build_object_geometry_table(
        plan=plan,
        labels=stage1["labels"],
    )

    enrich_df = enrich_object_table_streaming(
        plan=plan,
        object_df=geom_df,
        labels=stage1["labels"],
        probability_raster_path=prob_map[site_id],
        enrichment_layers=enrichment_layers,
        show_progress=show_progress,
    )

    site_summary = build_site_summary(
        plan=plan,
        cleaned_binary=stage1["cleaned_binary"],
        object_df=enrich_df,
    )

    object_csv = site_out / f"{site_id}__predicted_objects.csv"
    summary_json = site_out / f"{site_id}__site_summary.json"

    enrich_df.to_csv(object_csv, index=False)
    summary_json.write_text(json.dumps(site_summary, indent=2), encoding="utf-8")

    total_sec = time.perf_counter() - t0

    return {
        "site_id": site_id,
        "raw_binary_path": str(stage1["raw_binary_path"]),
        "cleaned_binary_path": str(stage1["cleaned_binary_path"]),
        "labels_path": str(stage1["labels_path"]),
        "object_csv_path": str(object_csv),
        "summary_json_path": str(summary_json),
        "object_df": enrich_df,
        "site_summary": site_summary,
        "runtime_seconds": total_sec,
    }
```

```python
def run_phase3_streaming_postprocessing(
    *,
    preflight_summary: dict,
    phase2_summary: dict,
    output_dir: str | Path,
    enrichment_layers: tuple[str, ...] = DEFAULT_POSTPROCESS_ENRICHMENT_LAYERS,
    show_progress: bool = True,
):
    test_sites = list(preflight_summary["split"]["test_sites"])
    results = []

    outer = tqdm(test_sites, desc="Phase 3 postprocessing", disable=not show_progress)
    for site_id in outer:
        outer.set_postfix({"site": site_id})
        res = run_phase3_postprocessing_for_site(
            preflight_summary=preflight_summary,
            phase2_summary=phase2_summary,
            site_id=site_id,
            output_dir=output_dir,
            enrichment_layers=enrichment_layers,
            show_progress=show_progress,
        )
        results.append({
            "site_id": site_id,
            "cleaned_binary_path": res["cleaned_binary_path"],
            "labels_path": res["labels_path"],
            "object_csv_path": res["object_csv_path"],
            "summary_json_path": res["summary_json_path"],
            "runtime_seconds": res["runtime_seconds"],
            "n_objects": int(len(res["object_df"])),
            "cover_fraction": float(res["site_summary"]["cover_fraction"]),
        })

    return {
        "results_df": pd.DataFrame(results),
        "per_site": {r["site_id"]: r for r in []},  # optional full map later
    }
```

```python
import math
import json
import time
from pathlib import Path

import numpy as np
import pandas as pd
from tqdm.auto import tqdm

import rasterio
from rasterio.enums import Resampling

from skimage import measure
from scipy.optimize import linear_sum_assignment

from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    jaccard_score,
)
```

```python
def phase3_result_map(phase3_summary: dict) -> dict[str, dict]:
    df = phase3_summary["results_df"].copy()
    out = {}
    for _, row in df.iterrows():
        out[str(row["site_id"])] = row.to_dict()
    return out

def load_phase3_site_outputs(site_id: str, phase3_summary: dict):
    row = phase3_result_map(phase3_summary)[site_id]

    cleaned_binary_path = Path(row["cleaned_binary_path"])
    labels_path = Path(row["labels_path"])
    object_csv_path = Path(row["object_csv_path"])
    summary_json_path = Path(row["summary_json_path"])

    with rasterio.open(cleaned_binary_path) as src:
        pred_binary = (src.read(1) > 0).astype(np.uint8)
        pred_profile = src.profile.copy()

    with rasterio.open(labels_path) as src:
        pred_labels = src.read(1).astype(np.int32)

    pred_objects_df = pd.read_csv(object_csv_path)
    pred_summary = json.loads(summary_json_path.read_text(encoding="utf-8"))

    return {
        "site_id": site_id,
        "pred_binary": pred_binary,
        "pred_labels": pred_labels,
        "pred_profile": pred_profile,
        "pred_objects_df": pred_objects_df,
        "pred_summary": pred_summary,
        "cleaned_binary_path": cleaned_binary_path,
        "labels_path": labels_path,
        "object_csv_path": object_csv_path,
        "summary_json_path": summary_json_path,
    }
```

```python
def safe_region_axis_major_length(r):
    if hasattr(r, "axis_major_length"):
        return float(r.axis_major_length)
    if hasattr(r, "major_axis_length"):
        return float(r.major_axis_length)
    return np.nan

def safe_region_axis_minor_length(r):
    if hasattr(r, "axis_minor_length"):
        return float(r.axis_minor_length)
    if hasattr(r, "minor_axis_length"):
        return float(r.minor_axis_length)
    return np.nan

def build_label_object_table_from_mask(
    *,
    plan: dict,
    label_mask: np.ndarray,
) -> pd.DataFrame:
    labels = measure.label(label_mask > 0, connectivity=2)
    regions = measure.regionprops(labels)

    pixel_area = canonical_pixel_area_m2(plan)
    transform = plan["canonical_grid"].transform

    rows = []
    for r in regions:
        cy, cx = r.centroid
        x, y = rasterio.transform.xy(transform, cy, cx, offset="center")

        perimeter_px = float(r.perimeter) if np.isfinite(r.perimeter) else 0.0
        area_px = int(r.area)
        compactness = float((4.0 * math.pi * area_px) / ((perimeter_px ** 2) + 1e-8))

        rows.append({
            "object_id": int(r.label),
            "area_px": area_px,
            "area_m2": float(area_px * pixel_area),
            "perimeter_px": perimeter_px,
            "centroid_row": float(cy),
            "centroid_col": float(cx),
            "centroid_x": float(x),
            "centroid_y": float(y),
            "bbox_min_row": int(r.bbox[0]),
            "bbox_min_col": int(r.bbox[1]),
            "bbox_max_row": int(r.bbox[2]),
            "bbox_max_col": int(r.bbox[3]),
            "eccentricity": float(getattr(r, "axis_eccentricity", getattr(r, "eccentricity", np.nan))),
            "solidity": float(getattr(r, "solidity", np.nan)),
            "extent": float(getattr(r, "extent", np.nan)),
            "equivalent_diameter": float(getattr(r, "equivalent_diameter_area", np.nan)),
            "major_axis_length": safe_region_axis_major_length(r),
            "minor_axis_length": safe_region_axis_minor_length(r),
            "compactness": compactness,
        })

    return pd.DataFrame(rows).sort_values("object_id").reset_index(drop=True), labels
```

```python
def resample_binary_mask(mask: np.ndarray, src_profile: dict, scale_factor: float) -> np.ndarray:
    if scale_factor == 1.0:
        return mask.astype(np.uint8)

    src_h, src_w = mask.shape
    dst_h = max(1, int(round(src_h / scale_factor)))
    dst_w = max(1, int(round(src_w / scale_factor)))

    dst = np.zeros((dst_h, dst_w), dtype=np.float32)

    src_transform = src_profile["transform"]
    dst_transform = src_transform * src_transform.scale(
        src_w / dst_w,
        src_h / dst_h,
    )

    reproject(
        source=mask.astype(np.float32),
        destination=dst,
        src_transform=src_transform,
        src_crs=src_profile["crs"],
        dst_transform=dst_transform,
        dst_crs=src_profile["crs"],
        resampling=Resampling.average,
    )

    return (dst >= 0.5).astype(np.uint8)

def meters_to_scale_factor(eval_resolution_m: float, native_resolution_m: float) -> float:
    return float(eval_resolution_m / native_resolution_m)
```

```python
def compute_raster_metrics(y_true: np.ndarray, y_pred: np.ndarray) -> dict:
    yt = y_true.reshape(-1).astype(np.uint8)
    yp = y_pred.reshape(-1).astype(np.uint8)

    intersection = float((yt * yp).sum())
    dice = float((2.0 * intersection) / (yt.sum() + yp.sum() + 1e-8))

    return {
        "dice": dice,
        "iou": float(jaccard_score(yt, yp, zero_division=0)),
        "precision": float(precision_score(yt, yp, zero_division=0)),
        "recall": float(recall_score(yt, yp, zero_division=0)),
        "cover_fraction_true": float(yt.mean()),
        "cover_fraction_pred": float(yp.mean()),
    }

def evaluate_raster_multiresolution(
    *,
    y_true: np.ndarray,
    y_pred: np.ndarray,
    profile: dict,
    resolutions_m: tuple[float, ...] = (1.0, 2.0, 5.0, 10.0),
):
    native_res_m = float(abs(profile["transform"].a))
    rows = []

    for res_m in resolutions_m:
        scale = meters_to_scale_factor(res_m, native_res_m)
        yt_r = resample_binary_mask(y_true, profile, scale)
        yp_r = resample_binary_mask(y_pred, profile, scale)
        metrics = compute_raster_metrics(yt_r, yp_r)
        metrics["resolution_m"] = float(res_m)
        rows.append(metrics)

    return pd.DataFrame(rows).sort_values("resolution_m").reset_index(drop=True)
```

```python
def mask_iou(a: np.ndarray, b: np.ndarray) -> float:
    inter = float(np.logical_and(a, b).sum())
    union = float(np.logical_or(a, b).sum())
    return inter / (union + 1e-8)

def build_object_masks(labels: np.ndarray) -> dict[int, np.ndarray]:
    ids = np.unique(labels)
    ids = ids[ids > 0]
    return {int(i): (labels == i) for i in ids}

def match_objects_by_iou(
    *,
    true_labels: np.ndarray,
    pred_labels: np.ndarray,
    iou_threshold: float = 0.3,
):
    true_masks = build_object_masks(true_labels)
    pred_masks = build_object_masks(pred_labels)

    true_ids = sorted(true_masks.keys())
    pred_ids = sorted(pred_masks.keys())

    if not true_ids or not pred_ids:
        return pd.DataFrame(columns=["true_object_id", "pred_object_id", "iou"])

    cost = np.ones((len(true_ids), len(pred_ids)), dtype=np.float32)

    for i, tid in enumerate(true_ids):
        for j, pid in enumerate(pred_ids):
            iou = mask_iou(true_masks[tid], pred_masks[pid])
            cost[i, j] = 1.0 - iou

    row_ind, col_ind = linear_sum_assignment(cost)

    rows = []
    for i, j in zip(row_ind, col_ind):
        iou = 1.0 - float(cost[i, j])
        if iou >= float(iou_threshold):
            rows.append({
                "true_object_id": int(true_ids[i]),
                "pred_object_id": int(pred_ids[j]),
                "iou": iou,
            })

    return pd.DataFrame(rows)

def compute_object_metrics(
    *,
    true_labels: np.ndarray,
    pred_labels: np.ndarray,
    true_objects_df: pd.DataFrame,
    pred_objects_df: pd.DataFrame,
    iou_threshold: float = 0.3,
):
    matches = match_objects_by_iou(
        true_labels=true_labels,
        pred_labels=pred_labels,
        iou_threshold=iou_threshold,
    )

    n_true = int(len(true_objects_df))
    n_pred = int(len(pred_objects_df))
    n_match = int(len(matches))

    precision = float(n_match / max(n_pred, 1))
    recall = float(n_match / max(n_true, 1))
    f1 = float((2 * precision * recall) / (precision + recall + 1e-8))

    count_error = float(n_pred - n_true)
    abs_count_error = abs(count_error)

    area_errors = []
    centroid_distances = []

    if not matches.empty:
        true_indexed = true_objects_df.set_index("object_id")
        pred_indexed = pred_objects_df.set_index("object_id")

        for _, row in matches.iterrows():
            tid = int(row["true_object_id"])
            pid = int(row["pred_object_id"])

            if tid in true_indexed.index and pid in pred_indexed.index:
                t = true_indexed.loc[tid]
                p = pred_indexed.loc[pid]

                area_errors.append(abs(float(p["area_m2"]) - float(t["area_m2"])))

                dr = float(p["centroid_row"]) - float(t["centroid_row"])
                dc = float(p["centroid_col"]) - float(t["centroid_col"])
                centroid_distances.append(float(np.sqrt(dr**2 + dc**2)))

    return {
        "object_precision": precision,
        "object_recall": recall,
        "object_f1": f1,
        "n_true_objects": n_true,
        "n_pred_objects": n_pred,
        "n_matched_objects": n_match,
        "count_error": count_error,
        "abs_count_error": float(abs_count_error),
        "mean_area_error_m2": float(np.mean(area_errors)) if area_errors else np.nan,
        "mean_centroid_distance_px": float(np.mean(centroid_distances)) if centroid_distances else np.nan,
    }
```

```python
def build_label_site_summary(
    *,
    plan: dict,
    label_mask: np.ndarray,
    label_objects_df: pd.DataFrame,
):
    pixel_area = canonical_pixel_area_m2(plan)
    site_area_m2 = float(label_mask.size * pixel_area)

    return {
        "cover_fraction": float(label_mask.mean()),
        "n_objects": int(len(label_objects_df)),
        "count_density_per_ha": float(len(label_objects_df) / max(site_area_m2 / 10000.0, 1e-8)),
        "mean_object_area_m2": float(label_objects_df["area_m2"].mean()) if not label_objects_df.empty else 0.0,
        "median_object_area_m2": float(label_objects_df["area_m2"].median()) if not label_objects_df.empty else 0.0,
    }

def compute_site_summary_metrics(label_summary: dict, pred_summary: dict) -> dict:
    keys = [
        "cover_fraction",
        "count_density_per_ha",
        "mean_object_area_m2",
        "median_object_area_m2",
    ]

    out = {}
    for k in keys:
        lv = float(label_summary.get(k, np.nan))
        pv = float(pred_summary.get(k, np.nan))
        out[f"{k}__label"] = lv
        out[f"{k}__pred"] = pv
        out[f"{k}__abs_error"] = abs(pv - lv) if np.isfinite(lv) and np.isfinite(pv) else np.nan

    return out
```

```python
def safe_unit_score_from_error(err: float, scale: float) -> float:
    if not np.isfinite(err):
        return 0.0
    return float(max(0.0, 1.0 - (err / max(scale, 1e-8))))

def compute_composite_score(
    *,
    raster_df: pd.DataFrame,
    object_metrics: dict,
    site_summary_metrics: dict,
    evaluation_cfg: dict,
):
    weights = evaluation_cfg["composite_weights"]

    raster_quality = float(raster_df["dice"].mean())

    object_quality_parts = [
        float(object_metrics["object_f1"]),
        safe_unit_score_from_error(float(object_metrics["abs_count_error"]), scale=max(object_metrics["n_true_objects"], 1)),
    ]
    if np.isfinite(object_metrics.get("mean_area_error_m2", np.nan)):
        object_quality_parts.append(safe_unit_score_from_error(float(object_metrics["mean_area_error_m2"]), scale=25.0))
    if np.isfinite(object_metrics.get("mean_centroid_distance_px", np.nan)):
        object_quality_parts.append(safe_unit_score_from_error(float(object_metrics["mean_centroid_distance_px"]), scale=25.0))
    object_quality = float(np.mean(object_quality_parts))

    site_summary_parts = []
    for k in [
        "cover_fraction__abs_error",
        "count_density_per_ha__abs_error",
        "mean_object_area_m2__abs_error",
        "median_object_area_m2__abs_error",
    ]:
        if np.isfinite(site_summary_metrics.get(k, np.nan)):
            # rough scales for first pass
            scale = 1.0 if "cover_fraction" in k else 50.0
            site_summary_parts.append(safe_unit_score_from_error(float(site_summary_metrics[k]), scale=scale))
    site_summary_score = float(np.mean(site_summary_parts)) if site_summary_parts else 0.0

    # For now generalization just mirrors the site-level overall quality.
    generalization = float(np.mean([raster_quality, object_quality, site_summary_score]))

    composite = (
        weights["raster_quality"] * raster_quality
        + weights["object_quality"] * object_quality
        + weights["site_summary"] * site_summary_score
        + weights["generalization"] * generalization
    )

    return {
        "raster_quality": raster_quality,
        "object_quality": object_quality,
        "site_summary_quality": site_summary_score,
        "generalization_quality": generalization,
        "composite_score": float(composite),
    }
```

```python
from rasterio.warp import reproject, Resampling

def reproject_label_mask_to_prediction_grid(
    *,
    label_bundle: dict,
    pred_profile: dict,
) -> np.ndarray:
    dst_h = int(pred_profile["height"])
    dst_w = int(pred_profile["width"])
    dst_transform = pred_profile["transform"]
    dst_crs = pred_profile["crs"]

    dst = np.zeros((dst_h, dst_w), dtype=np.uint8)

    with rasterio.open(label_bundle["binary_path"]) as src:
        reproject(
            source=rasterio.band(src, 1),
            destination=dst,
            src_transform=src.transform,
            src_crs=src.crs,
            src_nodata=src.nodata,
            dst_transform=dst_transform,
            dst_crs=dst_crs,
            dst_nodata=0,
            resampling=Resampling.nearest,
        )

    return (dst > 0).astype(np.uint8)

def run_phase4_evaluation_for_site(
    *,
    preflight_summary: dict,
    phase3_summary: dict,
    site_id: str,
):
    plan = preflight_summary["per_site_plans"][site_id]
    eval_cfg = preflight_summary["evaluation_config"]

    label_bundle = plan["label_bundle"]

    pred_bundle = load_phase3_site_outputs(site_id, phase3_summary)
    pred_binary = (pred_bundle["pred_binary"] > 0).astype(np.uint8)
    pred_labels = pred_bundle["pred_labels"].astype(np.int32)
    pred_objects_df = pred_bundle["pred_objects_df"].copy()

    # IMPORTANT: align label mask to prediction/canonical grid
    label_mask = reproject_label_mask_to_prediction_grid(
        label_bundle=label_bundle,
        pred_profile=pred_bundle["pred_profile"],
    )

    true_objects_df, true_labels = build_label_object_table_from_mask(
        plan=plan,
        label_mask=label_mask,
    )

    raster_df = evaluate_raster_multiresolution(
        y_true=label_mask,
        y_pred=pred_binary,
        profile=pred_bundle["pred_profile"],
        resolutions_m=tuple(eval_cfg["evaluate_resolutions_m"]),
    )

    object_metrics = compute_object_metrics(
        true_labels=true_labels,
        pred_labels=pred_labels,
        true_objects_df=true_objects_df,
        pred_objects_df=pred_objects_df,
        iou_threshold=float(eval_cfg["object_iou_threshold"]),
    )

    label_site_summary = build_label_site_summary(
        plan=plan,
        label_mask=label_mask,
        label_objects_df=true_objects_df,
    )
    site_summary_metrics = compute_site_summary_metrics(
        label_summary=label_site_summary,
        pred_summary=pred_bundle["pred_summary"],
    )

    composite = compute_composite_score(
        raster_df=raster_df,
        object_metrics=object_metrics,
        site_summary_metrics=site_summary_metrics,
        evaluation_cfg=eval_cfg,
    )

    return {
        "site_id": site_id,
        "raster_df": raster_df,
        "object_metrics": object_metrics,
        "label_site_summary": label_site_summary,
        "pred_site_summary": pred_bundle["pred_summary"],
        "site_summary_metrics": site_summary_metrics,
        "composite": composite,
    }
```

```python
def run_phase4_streaming_evaluation(
    *,
    preflight_summary: dict,
    phase3_summary: dict,
    show_progress: bool = True,
):
    test_sites = list(preflight_summary["split"]["test_sites"])
    per_site = {}
    rows = []

    outer = tqdm(test_sites, desc="Phase 4 evaluation", disable=not show_progress)

    for site_id in outer:
        outer.set_postfix({"site": site_id})
        res = run_phase4_evaluation_for_site(
            preflight_summary=preflight_summary,
            phase3_summary=phase3_summary,
            site_id=site_id,
        )
        per_site[site_id] = res

        row = {
            "site_id": site_id,
            **res["object_metrics"],
            **res["site_summary_metrics"],
            **res["composite"],
        }

        # add average raster stats
        raster_means = res["raster_df"].mean(numeric_only=True).to_dict()
        for k, v in raster_means.items():
            if k != "resolution_m":
                row[f"raster_{k}"] = float(v)

        rows.append(row)

    results_df = pd.DataFrame(rows).sort_values("site_id").reset_index(drop=True)
    return {
        "results_df": results_df,
        "per_site": per_site,
    }
```

```python
phase4_summary = run_phase4_streaming_evaluation(
    preflight_summary=phase1_preflight,
    phase3_summary=phase3_summary,
    show_progress=True,
)

display(phase4_summary["results_df"])
```

```python
import os

scripts_to_trace = [
    "/home/jovyan/work/Dry-shRub/shrub/Sprint 3/Base/IntELiMon_1_1_1.R",
    "/home/jovyan/work/Dry-shRub/shrub/Sprint 3/Base/IntELiMon_1_1_1_revised.R"
]

# R code to print the exact function calls leading up to a crash
traceback_code = 'options(error = function() { cat("\\n--- TRACEBACK ---\\n"); print(sys.calls()); q("no", status=1) })\n'

for script_path in scripts_to_trace:
    if os.path.exists(script_path):
        with open(script_path, 'r') as file:
            content = file.read()
        
        # Inject the traceback handler at the very top
        if "--- TRACEBACK ---" not in content:
            with open(script_path, 'w') as file:
                file.write(traceback_code + content)
            print(f"Added traceback handler to: {script_path}")
        else:
            print(f"Traceback handler already in: {script_path}")
    else:
        print(f"Could not find: {script_path}")
```

```python
import os

scripts_to_trace = [
    "/home/jovyan/work/Dry-shRub/shrub/Sprint 3/Base/IntELiMon_1_1_1.R",
    "/home/jovyan/work/Dry-shRub/shrub/Sprint 3/Base/IntELiMon_1_1_1_revised.R"
]

traceback_code = """
options(rlang_backtrace_on_error = "full")
options(error = function() {
    sink(stderr())
    cat("\\n================ TRACEBACK START ================\\n")
    print(sys.calls())
    if (requireNamespace("rlang", quietly = TRUE)) {
        cat("\\n--- RLANG TRACE ---\\n")
        print(rlang::trace_back(bottom = NULL))
    }
    cat("\\n=================================================\\n")
    sink()
    q("no", status=1)
})
"""

for script_path in scripts_to_trace:
    if os.path.exists(script_path):
        with open(script_path, 'r') as file:
            content = file.read()
            
        with open(script_path, 'w') as file:
            # We prepend it so the last options(error) defined wins
            file.write(traceback_code + content)
            
        print(f"Added stderr traceback handler to: {script_path}")
    else:
        print(f"Could not find: {script_path}")
```

```python
# Updated shim that handles the terra-to-raster conversion for legacy lidR
terra_bridge_shim = """
# --- Berkeley TLS Pipeline: Terra-to-Raster Bridge ---
if (!exists("tlsNormalize2")) {
  tlsNormalize2 <- function(las, dtm, min_res = 0.5, keep_ground = TRUE) {
    message("Using shimmed tlsNormalize2 (with Terra Bridge)...")
    
    # lidR v3 (legacy) only accepts 'RasterLayer' from the 'raster' package.
    # We must convert the modern 'SpatRaster' (terra) back to the old format.
    if (inherits(dtm, "SpatRaster")) {
      dtm <- raster::raster(dtm)
    }
    
    # Perform normalization using the converted DTM
    las <- lidR::lasnormalize(las, dtm)
    
    if (!keep_ground) {
      las <- lidR::filter_poi(las, Classification != 2)
    }
    if (min_res > 0) {
      las <- lidR::lasfilterdecimate(las, lidR::random(1/min_res))
    }
    return(las)
  }
}

# Ensure other shims also respect the requested package format
if (!exists("rasterize_terrain")) {
  rasterize_terrain <- function(las, res, algorithm = lidR::tin(), pkg = "terra", ...) {
    dtm <- lidR::grid_terrain(las, res = res, algorithm = algorithm, ...)
    if (pkg == "terra") return(terra::rast(dtm))
    return(dtm)
  }
}

if (!exists("rasterize_canopy")) {
  rasterize_canopy <- function(las, res, algorithm = lidR::p2r(), pkg = "terra", ...) {
    chm <- lidR::grid_canopy(las, res = res, algorithm = algorithm, ...)
    if (pkg == "terra") return(terra::rast(chm))
    return(chm)
  }
}
# ----------------------------------------------------
"""

scripts = [
    "/home/jovyan/work/Dry-shRub/shrub/Sprint 3/Base/IntELiMon_1_1_1.R",
    "/home/jovyan/work/Dry-shRub/shrub/Sprint 3/Base/IntELiMon_1_1_1_revised.R"
]

for script_path in scripts:
    with open(script_path, 'r') as f:
        content = f.read()
    
    # Remove all older shim versions
    import re
    content = re.sub(r"# --- Berkeley TLS.*?# ----------------------------------------------------", "", content, flags=re.DOTALL)
    content = re.sub(r"# --- lidR v3.*?# -------------------------------------------", "", content, flags=re.DOTALL)
    
    # Inject the bridge shim
    if "library(lidR)" in content:
        new_content = content.replace("library(lidR)", "library(lidR)" + terra_bridge_shim)
        with open(script_path, 'w') as f:
            f.write(new_content)
        print(f"Terra Bridge applied to: {script_path}")
```

```python
import os

scripts = [
    "/home/jovyan/work/Dry-shRub/shrub/Sprint 3/Base/IntELiMon_1_1_1.R",
    "/home/jovyan/work/Dry-shRub/shrub/Sprint 3/Base/IntELiMon_1_1_1_revised.R"
]

# The problematic line that triggers the defunct vctrs error
target_code = "dplyr::if_else(x == 0 & y == 0 & z == 0, 0, 1)"
# The safe replacement using base R
safe_replacement = "ifelse(x == 0 & y == 0 & z == 0, 0, 1)"

for script_path in scripts:
    if not os.path.exists(script_path):
        print(f"Skipping (not found): {script_path}")
        continue
        
    with open(script_path, 'r') as f:
        content = f.read()
    
    if target_code in content:
        new_content = content.replace(target_code, safe_replacement)
        with open(script_path, 'w') as f:
            f.write(new_content)
        print(f"Successfully patched defunct code in: {os.path.basename(script_path)}")
    else:
        # Check if it was already patched or has a slightly different format
        if "ifelse(x == 0 & y == 0 & z == 0, 0, 1)" in content:
             print(f"Code already patched in: {os.path.basename(script_path)}")
        else:
             print(f"Could not find target line in: {os.path.basename(script_path)}")

print("\nPatching complete. Please re-run your pipeline cell.")
```

```python
trial_record_min = {
    "trial_id": trial.trial_id,
    "created_at": trial.created_at,
    "status": trial.status,
    "labeling_config": trial.labeling_config,
    "features_config": trial.features_config,
    "modeling_config": trial.modeling_config,
    "postprocessing_config": trial.postprocessing_config,
    "score_summary": trial.score_summary,
}

registry.trials = [t for t in registry.trials if t["trial_id"] != trial.trial_id]
registry.trials.append(trial_record_min)

score = trial.score_summary.get("composite_score_placeholder")
if score is not None:
    if registry.best_score is None or score > registry.best_score:
        registry.best_score = score
        registry.best_trial_id = trial.trial_id

save_registry(registry)

registry
```
